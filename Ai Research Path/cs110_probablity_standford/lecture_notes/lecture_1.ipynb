## 1. Core Concepts & Definitions

### **Sample Space**
> **Definition:** The **sample space** (denoted by $S$) is the set of all possible outcomes of an experiment.

* An **experiment** is interpreted in the broadest possible sense—any procedure or situation with non-deterministic or variable outcomes.

---

### **Event**
> **Definition:** An **event** is a subset of the sample space ($A \subseteq S$).

* Probability uses **Set Theory** to represent outcomes and events rigorously:
  * **Sample Space ($S$):** The universe of outcomes.
  * **Event ($A$):** A specific collection or subset of outcomes within $S$.

---

### **Naive Definition of Probability**
> **Definition:** Let $S$ be a finite sample space with equally likely outcomes, and let $A \subseteq S$ be an event. The **naive probability** of $A$, denoted $P(A)$, is defined as:
> $$P(A) = \frac{|A|}{|S|} = \frac{\text{Number of favorable outcomes to } A}{\text{Total number of possible outcomes}}$$

#### **Crucial Assumptions & Limitations:**
1. **Finite Sample Space:** $|S| < \infty$
2. **Equally Likely Outcomes:** Every individual outcome in $S$ must have the exact same chance of occurring.

*If either assumption fails (e.g., continuous outcomes or biased events), the naive definition cannot be applied.*

## 2. Fundamental Rules of Counting

### **The Multiplication Rule**
> If an experiment consists of $r$ sequential stages/sub-experiments such that:
> 1. Stage 1 has $n_1$ possible outcomes,
> 2. Stage 2 has $n_2$ possible outcomes (regardless of the outcome of Stage 1),
> 3. $\dots$
> 4. Stage $r$ has $n_r$ possible outcomes,
> 
> then the total number of combined outcomes for the entire experiment is:
> $$\text{Total Outcomes} = n_1 \times n_2 \times \dots \times n_r$$

---

### **Binomial Coefficient**
> **Definition:** The number of ways to choose a subset of $k$ objects from a set of $n$ distinct objects (where order **does not** matter) is given by:
> $${n \choose k} = \frac{n!}{k!(n - k)!} \quad \text{for } 0 \le k \le n$$
> *Note: If $k > n$, ${n \choose k} = 0$.*

## 3. Sampling Table

When choosing $k$ objects out of $n$ total objects, the number of possible outcomes depends on whether **order matters** and whether **replacement is allowed**:

| | **Order Matters** | **Order Does Not Matter** |
| :--- | :--- | :--- |
| **With Replacement** | $n^k$ | ${n + k - 1 \choose k}$ |
| **Without Replacement** | $\frac{n!}{(n - k)!} = n(n - 1)\dots(n - k + 1)$ | ${n \choose k}$ |

## Derivations of the 4 Sampling Table Formulas

When sampling $k$ objects from $n$ distinct objects, the number of possible outcomes depends on whether **order matters** and whether **replacement is allowed**.

---

### 1. Order Matters, With Replacement ($n^k$)

**Problem:** Choose $k$ objects from $n$ distinct objects sequentially where order matters and replacement is allowed.

* **Step 1:** Consider $k$ ordered positions to fill: $(\text{Position}_1, \text{Position}_2, \dots, \text{Position}_k)$.
* **Step 2:** For the 1st position, there are $n$ possible choices.
* **Step 3:** Since replacement is allowed, the chosen object is returned to the set. Thus, for the 2nd position, there are still $n$ possible choices.
* **Step 4:** Repeating this process, every position from $1$ to $k$ has $n$ independent choices.
* **Step 5:** By the **Multiplication Rule**, the total number of distinct ordered sequences is:

$$\underbrace{n \times n \times \dots \times n}_{k \text{ times}} = n^k$$

---

### 2. Order Matters, Without Replacement ($\frac{n!}{(n-k)!}$)

**Problem:** Choose $k$ objects from $n$ distinct objects sequentially where order matters and each object can be chosen at most once.

* **Step 1:** Consider $k$ ordered positions to fill: $(\text{Position}_1, \text{Position}_2, \dots, \text{Position}_k)$.
* **Step 2:** For the 1st position, there are $n$ available choices.
* **Step 3:** Without replacement, the selected object cannot be reused. For the 2nd position, there are $(n - 1)$ choices remaining.
* **Step 4:** For the $i$-th position, there are $(n - i + 1)$ choices available, ending at $(n - k + 1)$ choices for the $k$-th position.
* **Step 5:** By the **Multiplication Rule**, the total number of ordered choices is:

$$n(n - 1)(n - 2) \dots (n - k + 1)$$

* **Step 6:** Multiplying and dividing by $(n - k)!$ yields:

$$\frac{n(n - 1)\dots(n - k + 1) \times (n - k)!}{(n - k)!} = \frac{n!}{(n - k)!}$$

---

### 3. Order Does Not Matter, Without Replacement ($\binom{n}{k}$)

**Problem:** Choose a subset of $k$ objects from $n$ distinct objects where order does not matter and each object can be chosen at most once.

* **Step 1:** Let $P$ be the number of ways to pick $k$ objects when **order matters** without replacement:

$$P = \frac{n!}{(n - k)!}$$

* **Step 2:** Any specific group of $k$ chosen objects can be arranged among themselves in $k!$ different orders.
* **Step 3:** Let $C$ be the number of ways to choose $k$ objects when **order does not matter**. Because each group of $k$ objects corresponds to $k!$ ordered sequences, we have:

$$P = C \times k!$$

* **Step 4:** Solving for $C$ gives the binomial coefficient:

$$C = \frac{P}{k!} = \frac{n!}{k!(n - k)!} = \binom{n}{k}$$

---

### 4. Order Does Not Matter, With Replacement ($\binom{n + k - 1}{k}$)

**Problem:** Choose $k$ objects from $n$ distinct categories with replacement where order does not matter (equivalently, distribute $k$ identical items into $n$ distinct bins).

* **Step 1:** Let $x_i$ be the number of items chosen from category $i$ (where $1 \le i \le n$). Any selection corresponds to an integer solution to:

$$x_1 + x_2 + \dots + x_n = k \quad \text{where } x_i \ge 0$$

* **Step 2 (Stars and Bars):** Represent the $k$ identical items using $k$ stars ($\star$). To separate the $n$ distinct categories, use $(n - 1)$ vertical bars ($|$).
* **Step 3:** A valid choice of items corresponds to a line of symbols consisting of $k$ stars and $(n - 1)$ bars. The total number of symbol positions is:

$$\text{Total positions} = k + (n - 1) = n + k - 1$$

* **Step 4:** Arranging the stars and bars is equivalent to choosing $k$ positions out of $(n + k - 1)$ total positions to place the stars.
* **Step 5:** By the definition of combinations, the total number of ways is:

$$\binom{n + k - 1}{k} \quad \text{or equivalently} \quad \binom{n + k - 1}{n - 1}$$

## 4. Worked Examples

### **Example 1: Coin Flips (Naive Probability)**
Flip a fair coin twice ($S = \{HH, HT, TH, TT\}$).  
What is the probability of getting two tails ($A = \{TT\}$)?

$$P(\text{Both Tails}) = \frac{|A|}{|S|} = \frac{1}{4} = 0.25$$

---

### **Example 2: Poker Hand (Full House)**
Find the probability of getting a **Full House** (3 cards of one rank and 2 cards of another rank) in a standard 5-card poker hand dealt from a 52-card deck.

1. **Total outcomes ($|S|$):** Choose 5 cards out of 52.
   $$|S| = {52 \choose 5}$$

2. **Favorable outcomes ($|A|$):**
   * Choose the rank for the triple: $13$ choices (${13 \choose 1}$)
   * Choose 3 cards of that rank: ${4 \choose 3}$ choices
   * Choose the rank for the pair: $12$ remaining choices (${12 \choose 1}$)
   * Choose 2 cards of that rank: ${4 \choose 2}$ choices

3. **Probability:**
   $$P(\text{Full House}) = \frac{13 \times {4 \choose 3} \times 12 \times {4 \choose 2}}{{52 \choose 5}}$$
