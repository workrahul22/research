# Fundamentals of Straight Lines in Machine Learning

A straight line is the most fundamental concept in Linear Algebra and Machine Learning (forming the foundation of **Linear Regression**) [00:00:00]. It provides a unique identifier for linear relationships between features [00:00:28].

---

### 1. Equation of a Line

The standard slope-intercept form of a straight line equation is:

$$y = mx + c$$

* **$y$:** Dependent Variable (Target/Output) [00:00:45].
* **$x$:** Independent Variable (Feature/Input) [00:00:45].
* **$m$:** **Slope** (Steepness/Gradient of the line) [00:00:51].
* **$c$:** **$y$-Intercept** (Value of $y$ when $x = 0$) [00:00:58].

> **Key Rule:** At least **2 points** are required to uniquely identify and draw a straight line [00:02:39].

---

### 2. Slope ($m$) Derivation

The slope represents how steep the line is relative to the $x$-axis [00:01:23]. Given two points $(x_1, y_1)$ and $(x_2, y_2)$ [00:02:05]:

$$m = \tan(\theta) = \frac{\text{Perpendicular}}{\text{Base}} = \frac{y_2 - y_1}{x_2 - x_1}$$ [00:03:46]

* **$\theta$:** Angle between the line and the horizontal $x$-axis [00:01:35].
* **$\Delta y = y_2 - y_1$:** Change in $y$ (Rise).
* **$\Delta x = x_2 - x_1$:** Change in $x$ (Run).

---

### 3. $y$-Intercept ($c$) & Derivation

* **$y$-intercept ($c$):** The point $(0, c)$ where the line intersects the $y$-axis [00:05:36].
* **Why $c$ is necessary:** Multiple parallel lines can share the exact same slope $m$. The intercept $c$ uniquely fixes the line's vertical position [00:04:39].

#### Proof / Derivation:
Using slope consistency between a general point $(x_1, y_1)$ and the intercept point $(0, c)$ [00:05:27]:

$$m = \frac{y_1 - c}{x_1 - 0}$$ [00:05:47]

$$m \cdot x_1 = y_1 - c$$ [00:06:09]

$$y_1 = m x_1 + c$$ [00:06:16]

Generalizing for any point $(x, y)$:
$$y = mx + c$$ [00:06:46]

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define 2 Points
x1, y1 = 2, 3
x2, y2 = 6, 11

# 2. Calculate Slope (m) and Intercept (c)
m = (y2 - y1) / (x2 - x1)
c = y1 - (m * x1)

print(f"Given Points: ({x1}, {y1}) and ({x2}, {y2})")
print(f"Calculated Slope (m): {m:.2f}")
print(f"Calculated Intercept (c): {c:.2f}")
print(f"Line Equation: y = {m:.2f}x + {c:.2f}")

# 3. Generate Line Data
x_vals = np.linspace(0, 8, 100)
y_vals = m * x_vals + c

# --- Plotting ---
plt.figure(figsize=(9, 6))

# Plot Straight Line
plt.plot(x_vals, y_vals, color='royalblue', linewidth=2.5, label=f'y = {m:.1f}x + {c:.1f}')

# Plot Sample Points
plt.scatter([x1, x2], [y1, y2], color='red', s=80, zorder=5, label=f'Points: ({x1},{y1}), ({x2},{y2})')
plt.scatter([0], [c], color='green', s=80, zorder=5, label=f'y-intercept (0, {c:.1f})')

# Plot Slope Triangle (Rise / Run)
plt.plot([x1, x2], [y1, y1], 'k--', alpha=0.6, label=f'Run (x2 - x1 = {x2-x1})')
plt.plot([x2, x2], [y1, y2], 'k:', alpha=0.8, label=f'Rise (y2 - y1 = {y2-y1})')

# Formatting
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.title("Straight Line: Slope (m) and Intercept (c) Visualized")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

# Coordinate Geometry for Machine Learning

Coordinate geometry forms the backbone of algorithms like **Support Vector Machines (SVM)** and **K-Nearest Neighbors (KNN)** [00:00:08].

---

### 1. Distance Between Two Points in 2D (KNN Core)
Given two points $P_1(x_1, y_1)$ and $P_2(x_2, y_2)$, the Euclidean distance between them using the **Pythagoras Theorem** is:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$ [00:02:09]

> **Machine Learning Link:** This Euclidean distance is directly used in **KNN** to find the nearest data points [00:15:04].

---

### 2. Parallel and Perpendicular Lines

#### **Parallel Lines:**
Lines with equations $y = mx + c_1$ and $y = mx + c_2$ have the **same slope ($m$)** because they form the same angle $\theta$ with the $x$-axis [00:03:54].
* **Condition:** $m_1 = m_2$ [00:04:14]

#### **Perpendicular Lines:**
If line 1 has slope $m_1$ ($\tan\theta$) and line 2 is perpendicular to it (forming angle $90^\circ + \theta$), then:

$$m_2 = \tan(90^\circ + \theta) = -\cot(\theta) = -\frac{1}{m_1}$$ [00:06:51]

* **Condition:** $m_1 \cdot m_2 = -1 \implies m_2 = -\frac{1}{m_1}$ [00:07:09]
* **Equation:** $y = -\frac{1}{m_1}x + c_2$ [00:07:25]

---

### 3. Distance Between Two Parallel Lines (SVM Margin Basis)
For two parallel lines $y = mx + c_1$ and $y = mx + c_2$, the shortest (perpendicular) distance $d$ between them is given by:

$$d = \frac{|c_2 - c_1|}{\sqrt{1 + m^2}}$$ [00:14:40]

> **Machine Learning Link:** This formula is directly used in **Support Vector Machines (SVM)** to maximize the hyper-plane margin [00:14:59].

---

### 📝 Practice Assignments Mentioned in Video
1. **3D Distance Formula:** Derive distance between $(x_1, y_1, z_1)$ and $(x_2, y_2, z_2)$ [00:08:27]:
   $$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$
2. **Standard Line Equation:** Convert parallel distance formula for $Ax + By + C_1 = 0$ and $Ax + By + C_2 = 0$ [00:15:11]:
   $$d = \frac{|C_2 - C_1|}{\sqrt{A^2 + B^2}}$$
3. **Point to Line Distance:** Shortest distance from point $(x_1, y_1)$ to line $Ax + By + C = 0$ [00:15:42]:
   $$d = \frac{|Ax_1 + By_1 + C|}{\sqrt{A^2 + B^2}}$$

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Distance between two points (KNN concept)
p1 = np.array([2, 3])
p2 = np.array([7, 9])
euclidean_dist = np.sqrt(np.sum((p2 - p1) ** 2))

# 2. Parallel lines & distance between them (SVM concept)
m = 1.5  # Slope
c1 = 2   # Line 1 intercept
c2 = 7   # Line 2 intercept

parallel_dist = abs(c2 - c1) / np.sqrt(1 + m**2)

print(f"Euclidean Distance between {p1} and {p2}: {euclidean_dist:.2f}")
print(f"Distance between parallel lines y = {m}x + {c1} and y = {m}x + {c2}: {parallel_dist:.2f}")

# --- Plotting ---
x = np.linspace(-2, 8, 100)
y1 = m * x + c1
y2 = m * x + c2

plt.figure(figsize=(9, 6))

# Plot Parallel Lines
plt.plot(x, y1, 'b-', label=f'Line 1: y = {m}x + {c1}')
plt.plot(x, y2, 'g-', label=f'Line 2: y = {m}x + {c2}')

# Plot Points
plt.scatter([p1[0], p2[0]], [p1[1], p2[1]], color='red', zorder=5)
plt.plot([p1[0], p2[0]], [p1[1], p2[1]], 'r--', label=f'Dist(P1, P2) = {euclidean_dist:.2f}')

# Formatting
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title("Coordinate Geometry Foundations for Machine Learning (KNN & SVM)")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

# Vectors in Machine Learning

Vectors form the foundation for multi-dimensional data representations in Machine Learning algorithms, Matrix operations, and Neural Networks [00:00:00].

---

### 1. Definition, Direction, and Magnitude

A vector $\mathbf{a}$ represents both **Magnitude** (length) and **Direction** [00:00:20].

#### **Direction:**
Expressed using unit vectors along coordinate axes ($\hat{i}, \hat{j}, \hat{k}$) [00:00:48]:
$$\mathbf{a} = a_x\hat{i} + a_y\hat{j} + a_z\hat{k}$$

#### **Magnitude (Euclidean Norm / $L_2$ Norm):**
The magnitude $\|\mathbf{a}\|$ represents the distance of the vector endpoint from the origin $(0,0,0)$ [00:02:18]:
$$\|\mathbf{a}\| = \sqrt{a_x^2 + a_y^2 + a_z^2}$$ [00:01:45]

---

### 2. Vector Addition (Resultant Vector)

When adding two vectors $\mathbf{a}$ and $\mathbf{b}$, we obtain a resultant vector $\mathbf{c} = \mathbf{a} + \mathbf{b}$ [00:03:06]:

$$\mathbf{a} + \mathbf{b} = (a_x + b_x)\hat{i} + (a_y + b_y)\hat{j} + (a_z + b_z)\hat{k}$$ [00:03:28]

* **Geometric Intuition:** Vector addition gives the **net resultant effect** of two forces or directions combining [00:04:55].

---

### 3. Vector Multiplication

#### **A. Dot Product (Scalar Product)**
The dot product projects one vector onto another to multiply their components in the same directional system [00:08:45].

$$\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \cos(\theta)$$ [00:06:05]

* **Projection of $\mathbf{a}$ on $\mathbf{b}$:** $\|\mathbf{a}\| \cos(\theta)$ [00:08:20]
* **Algebraic Form:** 
  $$\mathbf{a} \cdot \mathbf{b} = a_1 b_1 + a_2 b_2 + a_3 b_3$$ [00:09:40]
* **Machine Learning Link:** Core building block of matrix multiplication, dot-product attention, and linear transformations in **Neural Networks** [00:10:02].

#### **B. Cross Product (Vector Product)**
The cross product yields a third vector $\mathbf{c}$ that is **perpendicular (orthogonal)** to both $\mathbf{a}$ and $\mathbf{b}$ [00:10:37].

$$\mathbf{a} \times \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \sin(\theta) \mathbf{\hat{n}}$$ [00:11:30]

---

### 📝 Video Practice Assignment
* **Sine Component Intuition:** Find out why $\sin(\theta)$ appears in the cross product formula (Hint: It computes the perpendicular component / area of the parallelogram formed by the two vectors) [00:11:51].

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Vectors
a = np.array([4, 2])
b = np.array([1, 5])

# 2. Vector Addition (Resultant Vector)
c = a + b

# 3. Vector Magnitude
mag_a = np.linalg.norm(a)
mag_b = np.linalg.norm(b)

# 4. Dot Product Calculation
dot_product = np.dot(a, b)
cos_theta = dot_product / (mag_a * mag_b)
angle_rad = np.arccos(cos_theta)
angle_deg = np.degrees(angle_rad)

print(f"Vector A: {a}, Magnitude: {mag_a:.2f}")
print(f"Vector B: {b}, Magnitude: {mag_b:.2f}")
print(f"Resultant Vector C (A + B): {c}")
print(f"Dot Product (A . B): {dot_product}")
print(f"Angle between A and B: {angle_deg:.2f}°")

# --- Plotting ---
plt.figure(figsize=(8, 8))
origin = np.array([0, 0])

# Plot Vectors
plt.quiver(*origin, *a, angles='xy', scale_units='xy', scale=1, color='r', label='Vector A (4, 2)')
plt.quiver(*origin, *b, angles='xy', scale_units='xy', scale=1, color='g', label='Vector B (1, 5)')
plt.quiver(*origin, *c, angles='xy', scale_units='xy', scale=1, color='b', label='Resultant Vector C (A+B)')

# Vector Addition Parallelogram (dashed lines)
plt.plot([a[0], c[0]], [a[1], c[1]], 'g--', alpha=0.6)
plt.plot([b[0], c[0]], [b[1], c[1]], 'r--', alpha=0.6)

# Formatting
plt.xlim(-1, 7)
plt.ylim(-1, 8)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title("Vector Addition & Projection Visualizer")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

# Matrices in Machine Learning

Matrices are 2D arrays of numbers arranged in rows and columns [00:00:00]. They form the backbone of dataset representation, image processing, CPU/GPU rendering, and Neural Networks [00:01:33].

---

### 1. Matrix Notation & Addition

A matrix $A$ with $m$ rows and $n$ columns has dimensions $m \times n$ [00:01:07].

#### **Element-wise Addition / Subtraction:**
Matrices can only be added or subtracted if they have identical dimensions ($m \times n$) [00:02:29].
$$C_{ij} = A_{ij} + B_{ij}$$ [00:02:16]

#### **Scalar Multiplication:**
Multiplying a matrix by a constant scales every individual element inside the matrix [00:03:48].
$$k \cdot A_{ij} = (k \cdot A)_{ij}$$ [00:03:54]

---

### 2. Matrix Multiplication ($A \times B$)

To multiply matrix $A$ ($m \times n$) by matrix $B$ ($p \times q$), the inner dimensions must match [00:07:36]:

$$\text{Condition: } n = p \quad \text{(Columns of } A = \text{Rows of } B\text{)}$$ [00:07:43]

#### **Element Calculation:**
The entry at row $i$ and column $j$ of the resulting matrix $C$ is computed by taking the dot product of row $i$ of $A$ and column $j$ of $B$ [00:04:32]:

$$C_{ij} = \sum_{k} A_{ik} B_{kj}$$ [00:05:00]

> **Video Assignment Answer:** If $A$ is $(m \times n)$ and $B$ is $(n \times q)$, the resulting matrix $C$ has dimensions **$m \times q$** [00:08:13].

---

### 3. Matrix Transpose ($A^T$)

Transposing a matrix swaps its rows and columns ($A_{ij}^T = A_{ji}$) [00:09:11].
* Dimensions change from $m \times n \implies n \times m$ [00:09:46].

#### **Application in Neural Networks (Dot Product via Matrix Multiplication):**
To calculate the dot product of two row vectors $u$ and $v$ using matrix operations, transpose one of them into a column vector [00:10:57]:

$$u \cdot v = u \cdot v^T$$ [00:11:04]

* **Machine Learning Link:** This exact $W X^T + b$ operation is used for layer forward passes in **Deep Learning / Neural Networks** [00:11:36].

```python
import numpy as np

# 1. Define Matrices
A = np.array([
    [2, 1],
    [3, 4]
])

B = np.array([
    [6, 2],
    [1, 8]
])

# 2. Matrix Addition & Scalar Multiplication
C_add = A + B
C_scalar = 3 * A

print("--- Matrix Addition (A + B) ---")
print(C_add)

print("\n--- Scalar Multiplication (3 * A) ---")
print(C_scalar)

# 3. Matrix Multiplication (A x B)
C_mult = np.dot(A, B)
print("\n--- Matrix Multiplication (A x B) ---")
print(C_mult)

# 4. Matrix Transpose
A_T = A.T
print("\n--- Transpose of A (A^T) ---")
print(A_T)

# 5. Vector Dot Product using Transpose (Neural Network concept)
u = np.array([[3, 1, 2]])  # Shape (1, 3)
v = np.array([[4, 1, 3]])  # Shape (1, 3)

# Transpose v to shape (3, 1) for matrix multiplication
dot_product_matrix = np.dot(u, v.T)

print("\n--- Vector Dot Product via Matrix Transpose (u . v^T) ---")
print(f"Result: {dot_product_matrix[0][0]}")
```

# Determinants, Eigenvalues, and Eigenvectors

Determinants and Eigen-decomposition form the mathematical foundations for dimensionality reduction techniques like **Principal Component Analysis (PCA)** and stability analysis in systems [00:00:00].

---

### 1. Determinants & System of Equations

The determinant $|A|$ of a matrix measures scaling factor and solvability [00:00:17].

#### **Determinant Calculation ($3 \times 3$ Matrix):**
Using Minors ($M_{ij}$) and Cofactors ($C_{ij} = (-1)^{i+j} M_{ij}$) [00:02:47]:
$$|A| = a_{11} C_{11} + a_{12} C_{12} + a_{13} C_{13}$$ [00:04:06]

#### **Machine Learning Intuition:**
* **$|A| \neq 0$:** The system of equations has a **unique solution** (non-parallel lines / non-coplanar planes intersecting at a single point) [00:06:05].
* **$|A| = 0$:** Singular matrix (dependent features/linear dependence).

---

### 2. Eigenvalues ($\lambda$) and Eigenvectors ($v$)

An **Eigenvector** is a special non-zero vector $x$ that, when transformed by matrix $A$, does **not change its direction**—it only scales by factor $\lambda$ (the **Eigenvalue**) [00:09:19].

$$A x = \lambda x$$ [00:08:06]

* **$A$:** Square Transformation Matrix [00:08:29].
* **$x$:** Eigenvector (Direction remains invariant) [00:08:37].
* **$\lambda$:** Eigenvalue (Scalar magnitude scaling factor) [00:08:47].

#### **Characteristic Equation:**
To find eigenvalues, set the determinant of $(A - \lambda I)$ to zero [00:10:24]:

$$|A - \lambda I| = 0$$ [00:10:58]

Where $I$ is the Identity Matrix [00:10:33].

> **Machine Learning Link:** Eigenvectors correspond to the directions of maximum variance in data (**Principal Components in PCA**) [00:07:36].

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Transformation Matrix
A = np.array([
    [4, 1],
    [2, 3]
])

# 2. Determinant Calculation
det_A = np.linalg.det(A)
print(f"Determinant |A|: {det_A:.2f}")

# 3. Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("\n--- Eigen Decomposition ---")
for i in range(len(eigenvalues)):
    print(f"Eigenvalue {i+1} (λ): {eigenvalues[i]:.2f}")
    print(f"Eigenvector {i+1} (v): {eigenvectors[:, i]}\n")

# Verify A * v = λ * v for Eigenvector 1
v1 = eigenvectors[:, 0]
lambda1 = eigenvalues[0]

Av1 = np.dot(A, v1)
lambda_v1 = lambda1 * v1

print("--- Verification: A * v1 == λ1 * v1 ---")
print(f"A * v1   : {Av1}")
print(f"λ1 * v1  : {lambda_v1}")

# --- Visualization ---
origin = np.array([0, 0])
plt.figure(figsize=(7, 7))

# Plot Eigenvectors (Directions that do not rotate under transformation)
plt.quiver(*origin, *eigenvectors[:, 0], color='r', scale=3, label=f'Eigenvector 1 (λ={eigenvalues[0]:.1f})')
plt.quiver(*origin, *eigenvectors[:, 1], color='g', scale=3, label=f'Eigenvector 2 (λ={eigenvalues[1]:.1f})')

plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title("Eigenvectors Visualized (Directions invariant under Transformation)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
