# Calculus for Machine Learning: Functions, Domain, Range & Plotting

Calculus deals with continuous change in functions [00:00:06]. In Machine Learning, calculus is fundamental for optimizing models (e.g., Gradient Descent).

---

### 1. What is a Function?
A function $f(x)$ takes an input $x$, applies an operation, and returns an output $y$ [00:00:43]:
$$y = f(x)$$ [00:00:55]

#### **Core Concepts:**
* **Domain:** Range of input values $x$ that can be fed into the function [00:02:32].
* **Range:** Range of output values $y$ produced by the function [00:03:03].

#### **Example: $y = \sqrt{x}$**
* **Domain:** $[0, \infty)$ — Negative values inside square roots are invalid for real numbers [00:07:36].
* **Range:** $[0, \infty)$ — Principal square roots output non-negative real numbers [00:09:33].

---

### 2. Vertical Line Test for Valid Functions
A graph represents a valid function if **no vertical line** intersects the graph at more than one point [00:10:21].
* **Valid Function ($f_1$):** Each input $x$ maps to exactly one output $y$ [00:10:49].
* **Invalid Function ($f_2$):** A single $x$ maps to multiple $y$ values [00:11:38].

---

### 3. Function Plotting (Assignment: $y = \sqrt{x}$)
Plotting helps visualize gradient behavior, inflection points, and model optimization surfaces [00:05:30].

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Function Domain (x >= 0 for square root)
x = np.linspace(0, 10, 500)
y = np.sqrt(x)

# 2. Plotting Function: y = sqrt(x)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r'$y = \sqrt{x}$', color='b', linewidth=2)

# 3. Vertical Line Test Demonstration (x = 4)
x_test = 4
y_test = np.sqrt(x_test)

# Plot vertical line at x = 4
plt.axvline(x=x_test, color='r', linestyle='--', label=f'Vertical Line Test at x={x_test}')
plt.plot(x_test, y_test, 'ro', markersize=8) # Intersection point

# Labels & Aesthetics
plt.title(r"Function Plotting & Vertical Line Test ($y = \sqrt{x}$)", fontsize=12)
plt.xlabel("Domain (x >= 0)")
plt.ylabel("Range (y >= 0)")
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

# Nested Functions & Function Transformations

Function transformations and compositions are vital for understanding feature engineering, activation functions, and neural network layer propagation in Machine Learning [00:00:12].

---

### 1. Nested (Composite) Functions: $f(g(x))$
A composite function applies one function to the result of another [00:00:23].
* **Example:** Given $g(x) = x + 1$ and $f(x) = \sqrt{x}$, the composite function is [00:02:49]:
  $$f(g(x)) = \sqrt{x + 1}$$ [00:03:39]
* **Domain Consideration:** The domain shrinks because $x + 1 \ge 0 \implies x \ge -1$. Thus, Domain = $[-1, \infty)$ [00:04:36].

---

### 2. Vertical Scaling & Reflection ($c \cdot f(x)$)
Multiplying a function by a scalar $c$ scales its range vertically along the y-axis [00:05:35]:
* **$c > 1$:** Vertically stretches the function [00:07:04].
* **$0 < c < 1$:** Vertically compresses the function [00:09:19].
* **$c < 0$:** Reflects the function vertically across the x-axis (mirror image) [00:11:30].

---

### 3. Horizontal Scaling & Compression ($f(c \cdot x)$)
Multiplying the input variable $x$ by $c$ affects the frequency and width along the x-axis [00:13:12]:
* **$c > 1$:** Horizontally squashes/compresses the function toward the y-axis [00:17:47].
* **$0 < c < 1$:** Horizontally stretches the function away from the y-axis.

---

### 4. Horizontal & Vertical Shifts
* **Horizontal Translation ($f(x + c)$):** Adding $c$ inside the argument shifts the graph **left** by $c$ units; subtracting $c$ shifts it **right** [00:19:21].
* **Vertical Translation ($f(x) + c$):** Adding $c$ outside shifts the entire graph **up** by $c$ units; subtracting $c$ shifts it **down** [00:25:14].

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define x values
x = np.linspace(-2*np.pi, 2*np.pi, 1000)

# 2. Transformations of sin(x)
y_base = np.sin(x)
y_vert_scaled = 2 * np.sin(x)          # Vertical stretch (2 * f(x))
y_horiz_squashed = np.sin(2 * x)       # Horizontal squash (f(2x))
y_reflected = -1 * np.sin(x)           # Reflection (-f(x))

# 3. Setup Plotting Subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Subplot 1: Vertical Scaling
axes[0, 0].plot(x, y_base, label=r'$y = \sin(x)$', color='blue', linestyle='--')
axes[0, 0].plot(x, y_vert_scaled, label=r'$y = 2\sin(x)$', color='red')
axes[0, 0].set_title(r"Vertical Scaling ($c \cdot f(x)$)")
axes[0, 0].grid(True, linestyle="--", alpha=0.5)
axes[0, 0].legend()

# Subplot 2: Horizontal Compression
axes[0, 1].plot(x, y_base, label=r'$y = \sin(x)$', color='blue', linestyle='--')
axes[0, 1].plot(x, y_horiz_squashed, label=r'$y = \sin(2x)$', color='green')
axes[0, 1].set_title(r"Horizontal Compression ($f(c \cdot x)$)")
axes[0, 1].grid(True, linestyle="--", alpha=0.5)
axes[0, 1].legend()

# Subplot 3: Reflection across X-axis
axes[1, 0].plot(x, y_base, label=r'$y = \sin(x)$', color='blue', linestyle='--')
axes[1, 0].plot(x, y_reflected, label=r'$y = -\sin(x)$', color='orange')
axes[1, 0].set_title(r"Vertical Reflection ($-f(x)$)")
axes[1, 0].grid(True, linestyle="--", alpha=0.5)
axes[1, 0].legend()

# Subplot 4: Parabola Shift f(x+c) vs f(x)+c
x_parabola = np.linspace(-4, 4, 400)
y_orig = x_parabola**2
y_shift_left = (x_parabola + 2)**2
y_shift_up = x_parabola**2 + 2

axes[1, 1].plot(x_parabola, y_orig, label=r'$y = x^2$', color='blue', linestyle='--')
axes[1, 1].plot(x_parabola, y_shift_left, label=r'$y = (x+2)^2$ (Left Shift)', color='purple')
axes[1, 1].plot(x_parabola, y_shift_up, label=r'$y = x^2 + 2$ (Up Shift)', color='brown')
axes[1, 1].set_title(r"Translations: $f(x+c)$ vs $f(x) + c$")
axes[1, 1].set_ylim(-1, 10)
axes[1, 1].grid(True, linestyle="--", alpha=0.5)
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

# Differentiation for Machine Learning

Differentiation measures the rate of change of a function with respect to an input variable [00:01:41]. Geometrically, the derivative at a point represents the **slope of the tangent line** at that specific point on a curve [00:01:12].

---

### 1. Geometric Interpretation
* **Notations:** Expressed as $\frac{dy}{dx}$ or $f'(x)$ [00:00:10].
* **Tangent & Slope:** At a specific point $x_0$, the derivative gives $\tan(\theta)$, where $\theta$ is the angle the tangent line makes with the positive x-axis [00:01:50].
* **Machine Learning Context:** Derivatives form the foundation of optimization algorithms like Gradient Descent to minimize cost functions [00:11:21].

---

### 2. First Principles of Differentiation (Limit Definition)
To find the slope at a single point, we take a small neighboring point $x + h$ and evaluate the secant line slope as $h \to 0$ [00:04:33]:

$$\frac{dy}{dx} = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$ [00:06:24]

#### Derivation Example: $f(x) = x^2$ [00:08:05]
1. Substitute into the formula:
   $$\frac{dy}{dx} = \lim_{h \to 0} \frac{(x + h)^2 - x^2}{h}$$ [00:08:14]
2. Expand the numerator:
   $$= \lim_{h \to 0} \frac{x^2 + 2xh + h^2 - x^2}{h} = \lim_{h \to 0} \frac{h(2x + h)}{h}$$ [00:08:35]
3. Cancel $h$ and evaluate the limit as $h \to 0$:
   $$= \lim_{h \to 0} (2x + h) = 2x$$ [00:09:14]

---

### 3. Common Derivative Rules
* **Power Rule:** $\frac{d}{dx}[x^n] = n \cdot x^{n-1}$ [00:10:56]
* **Sine:** $\frac{d}{dx}[\sin(x)] = \cos(x)$ [00:11:28]
* **Cosine:** $\frac{d}{dx}[\cos(x)] = -\sin(x)$ [00:11:38]

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the function and its analytical derivative
def f(x):
    return x**2

def df(x):
    return 2*x  # Derivative of x^2 is 2x

# 2. Setup domain and point of interest
x_vals = np.linspace(-3, 3, 400)
y_vals = f(x_vals)

x0 = 1.5      # Point where we calculate the tangent
y0 = f(x0)
slope = df(x0) # Slope at x0 = 3.0

# 3. Tangent line equation: y - y0 = m(x - x0) -> y = m(x - x0) + y0
x_tangent = np.linspace(0.5, 2.5, 100)
y_tangent = slope * (x_tangent - x0) + y0

# 4. Plotting
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, label=r'$f(x) = x^2$', color='blue', linewidth=2)
plt.plot(x_tangent, y_tangent, label=f'Tangent at x={x0} (Slope = {slope:.1f})', color='red', linestyle='--')
plt.scatter([x0], [y0], color='red', zorder=5, s=60, label=f'Point ({x0}, {y0})')

# Annotate Slope
plt.title(r'Differentiation: $f(x) = x^2$ and Tangent Line Slope $\frac{dy}{dx} = 2x$', fontsize=12)
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(0, color='black', linewidth=0.8, linestyle=':')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

# Differentiation Rules for Machine Learning

When working with deep neural networks and complex loss functions, basic derivative rules are combined to differentiate composite and combined functions [00:00:09].

---

### 1. Product Rule
Used when differentiating the product of two functions $f(x) \cdot g(x)$ [00:00:12]:

$$\frac{d}{dx} \big[ f(x) \cdot g(x) \big] = f(x) \cdot g'(x) + f'(x) \cdot g(x)$$ [00:00:36]

* **Example:** Differentiate $y = x^2 \sin(x)$ [00:01:20]:
  $$\frac{dy}{dx} = x^2 \cdot \cos(x) + 2x \cdot \sin(x)$$ [00:01:52]

---

### 2. Quotient Rule
Used when differentiating the ratio of two functions $\frac{f(x)}{g(x)}$ [00:02:10]:

$$\frac{d}{dx} \left[ \frac{f(x)}{g(x)} \right] = \frac{f'(x) \cdot g(x) - f(x) \cdot g'(x)}{[g(x)]^2}$$ [00:02:41]

* **Example:** Differentiate $y = \frac{x^2}{\sin(x)}$ [00:02:10]:
  $$\frac{dy}{dx} = \frac{2x \sin(x) - x^2 \cos(x)}{\sin^2(x)}$$ [00:03:32]

---

### 3. Chain Rule
Essential for calculating gradients during backpropagation in Deep Learning when differentiating composite functions $f(g(x))$ [00:03:55]:

$$\frac{d}{dx} \big[ f(g(x)) \big] = f'(g(x)) \cdot g'(x)$$ [00:04:16]

* **Nested Composite Example:** Differentiate $y = e^{\sin(x^2)}$ [00:05:07]:
  $$\frac{dy}{dx} = 2x \cdot \cos(x^2) \cdot e^{\sin(x^2)}$$ [00:05:42]

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define functions and analytical derivatives
def f_product(x):
    return x**2 * np.sin(x)

def df_product(x):
    return x**2 * np.cos(x) + 2*x * np.sin(x)

def f_chain(x):
    return np.exp(np.sin(x**2))

def df_chain(x):
    return 2*x * np.cos(x**2) * np.exp(np.sin(x**2))

# 2. Generate x values
x_vals = np.linspace(-2.5, 2.5, 500)

# 3. Create subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Subplot 1: Product Rule Visualization
axes[0].plot(x_vals, f_product(x_vals), label=r'$y = x^2 \sin(x)$', color='blue', linewidth=2)
axes[0].plot(x_vals, df_product(x_vals), label=r"$\frac{dy}{dx} = x^2 \cos(x) + 2x \sin(x)$", color='red', linestyle='--')
axes[0].set_title("Product Rule: Function vs Derivative")
axes[0].grid(True, linestyle="--", alpha=0.5)
axes[0].legend()

# Subplot 2: Chain Rule Visualization
axes[1].plot(x_vals, f_chain(x_vals), label=r'$y = e^{\sin(x^2)}$', color='green', linewidth=2)
axes[1].plot(x_vals, df_chain(x_vals), label=r"$\frac{dy}{dx} = 2x \cos(x^2) e^{\sin(x^2)}$", color='orange', linestyle='--')
axes[1].set_title("Chain Rule: Function vs Derivative")
axes[1].grid(True, linestyle="--", alpha=0.5)
axes[1].legend()

plt.tight_layout()
plt.show()
```

# Applications of Differentiation in Machine Learning

Derivatives represent the slope (rate of change) of a function and help us analyze function behavior, find turning points (inflection points), and compute minima/maxima for optimization algorithms [00:06:37].

---

### 1. Curve Sketching & Finding Stationary Points
Instead of plotting individual points—which fails to capture accurate curve behavior [00:05:05]—we set the first derivative to zero to find where the slope is flat ($f'(x) = 0$) [00:07:07].

* **Given Function:** 
  $$y = x + x^2 - x^3$$ [00:00:28]

* **First Derivative (Slope):**
  $$\frac{dy}{dx} = 1 + 2x - 3x^2$$ [00:06:44]

* **Finding Stationary Points ($dy/dx = 0$):**
  $$3x^2 - 2x - 1 = 0 \implies (x - 1)(3x + 1) = 0$$ [00:07:36]
  * $x_1 = 1 \implies y_1 = 1$ [00:08:24]
  * $x_2 = -\frac{1}{3} \implies y_2 \approx -0.185$ ($-\frac{5}{27}$) [00:09:46]

* **Inflection/Turning Points:** The graph changes direction at $(1, 1)$ and $(-\frac{1}{3}, -0.185)$ [00:15:25].

---

### 2. Local vs. Global Minima/Maxima

* **Local Minima:** The point where a function reaches its lowest value within a specific bounded interval/neighborhood [00:17:25]. 
  * For $y = x + x^2 - x^3$ in the local region around $x \in [-1, 1]$, the local minimum occurs at $x = -\frac{1}{3}$ [00:19:06].
* **Global Minima:** The point where a function reaches its absolute lowest value across its entire domain [00:19:19].
  * For $y = x + x^2 - x^3$, as $x \to \infty$, $y \to -\infty$, meaning the global minimum is $-\infty$ [00:19:36].
  * For a simple quadratic loss function $y = x^2$, the point $(0, 0)$ serves as both the local and global minimum [00:20:09].

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the function and its first derivative
def f(x):
    return x + x**2 - x**3

def df(x):
    return 1 + 2*x - 3*x**2

# 2. Critical stationary points where df/dx = 0
x_crit = np.array([1.0, -1/3])
y_crit = f(x_crit)

# 3. Generate domain points
x_vals = np.linspace(-1.5, 1.8, 500)
y_vals = f(x_vals)

# 4. Plot the curve and critical points
plt.figure(figsize=(10, 6))

# Main curve and derivative
plt.plot(x_vals, y_vals, label=r'$y = x + x^2 - x^3$', color='indigo', linewidth=2.5)
plt.plot(x_vals, df(x_vals), label=r"$\frac{dy}{dx} = 1 + 2x - 3x^2$", color='crimson', linestyle='--', alpha=0.7)

# Highlight stationary points (slope = 0)
plt.scatter(x_crit, y_crit, color='orange', zorder=5, s=80, label=r'Stationary Points ($\frac{dy}{dx} = 0$)')

# Annotations
plt.annotate(f'Local Max\n(1.0, 1.0)', xy=(1, 1), xytext=(1.1, 1.2),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))
plt.annotate(f'Local Min\n(-0.33, -0.18)', xy=(-1/3, f(-1/3)), xytext=(-0.8, -0.8),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

plt.axhline(0, color='black', linewidth=0.8, linestyle=':')
plt.axvline(0, color='black', linewidth=0.8, linestyle=':')
plt.title("Applications of Differentiation: Curve Behavior & Critical Points", fontsize=12)
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

plt.show()
```
