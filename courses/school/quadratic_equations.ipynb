## 1. Definition and Standard Form

A **quadratic equation** in one variable $x$ is an equation that can be written as

$$ax^2 + bx + c = 0, \qquad a \neq 0$$

where $a, b, c$ are constants (usually real numbers).

| Term | Name |
|------|------|
| $ax^2$ | quadratic (second-degree) term; $a$ is the **leading coefficient** |
| $bx$ | linear term |
| $c$ | constant term |

- The **degree** is 2 (highest power of $x$).
- If $a = 0$, the equation becomes $bx + c = 0$, which is **linear**, not quadratic.
- $b$ or $c$ (or both) may be zero:
  - $ax^2 = 0$ (pure, $b = c = 0$)
  - $ax^2 + c = 0$ (pure quadratic, $b = 0$)
  - $ax^2 + bx = 0$ (no constant term, $c = 0$)

**Checking whether an equation is quadratic:** always simplify first.

- $(x+2)^2 = x^2 + 5$ → $x^2 + 4x + 4 = x^2 + 5$ → $4x - 1 = 0$. **Not** quadratic (the $x^2$ terms cancel).
- $x(x+1) = 6$ → $x^2 + x - 6 = 0$. Quadratic.

## 2. Roots (Solutions) of a Quadratic

A number $\alpha$ is a **root** (or **zero**, or **solution**) of $ax^2 + bx + c = 0$ if

$$a\alpha^2 + b\alpha + c = 0.$$

Key facts:
- A quadratic equation has **exactly two roots** (counting repetition, and allowing complex numbers).
- Over the reals it has **0, 1 (repeated), or 2** distinct real roots.
- The real roots are the $x$-intercepts of the parabola $y = ax^2 + bx + c$.
- If $\alpha$ and $\beta$ are the roots, then $ax^2 + bx + c = a(x - \alpha)(x - \beta)$ (**factor theorem**).

## 3. Methods of Solving

### 3.1 Factorisation (Splitting the Middle Term)

To factor $ax^2 + bx + c$:
1. Compute the product $a \cdot c$.
2. Find two numbers $p, q$ with $p + q = b$ and $p \cdot q = ac$.
3. Rewrite $bx = px + qx$ and factor by grouping.
4. Use the **zero-product property**: if $AB = 0$ then $A = 0$ or $B = 0$.

**Example.** Solve $6x^2 - 5x - 6 = 0$.

- $ac = -36$; need $p + q = -5$, $pq = -36$ → $p = -9,\ q = 4$.
- $6x^2 - 9x + 4x - 6 = 3x(2x - 3) + 2(2x - 3) = (3x + 2)(2x - 3)$
- $x = -\tfrac{2}{3}$ or $x = \tfrac{3}{2}$.

**Special cases**
- $ax^2 + bx = 0$ → $x(ax + b) = 0$ → $x = 0$ or $x = -\tfrac{b}{a}$.
  *Never divide both sides by $x$ — you would lose the root $x = 0$.*
- $ax^2 + c = 0$ → $x^2 = -\tfrac{c}{a}$ → $x = \pm\sqrt{-\tfrac{c}{a}}$.
- Difference of squares: $x^2 - k^2 = (x - k)(x + k)$.
- Perfect square: $x^2 \pm 2kx + k^2 = (x \pm k)^2$.

### 3.2 Completing the Square

Idea: rewrite the equation as $(x + h)^2 = k$, then take square roots.

Steps:
1. Divide by $a$ so the coefficient of $x^2$ is 1.
2. Move the constant to the right-hand side.
3. Add $\left(\tfrac{\text{coefficient of } x}{2}\right)^2$ to **both** sides.
4. Write the left side as a perfect square and take $\pm\sqrt{\ }$.

**Example.** Solve $2x^2 - 8x + 3 = 0$.

$$x^2 - 4x + \tfrac{3}{2} = 0$$
$$x^2 - 4x = -\tfrac{3}{2}$$
$$x^2 - 4x + 4 = 4 - \tfrac{3}{2} = \tfrac{5}{2}$$
$$(x - 2)^2 = \tfrac{5}{2}$$
$$x = 2 \pm \sqrt{\tfrac{5}{2}} = 2 \pm \tfrac{\sqrt{10}}{2}$$

### 3.3 The Quadratic Formula (Sridharacharya's Formula)

$$\boxed{x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}}$$

**Derivation** (completing the square on the general equation):

$$ax^2 + bx + c = 0$$
$$x^2 + \frac{b}{a}x = -\frac{c}{a}$$
$$x^2 + \frac{b}{a}x + \frac{b^2}{4a^2} = \frac{b^2}{4a^2} - \frac{c}{a} = \frac{b^2 - 4ac}{4a^2}$$
$$\left(x + \frac{b}{2a}\right)^2 = \frac{b^2 - 4ac}{4a^2}$$
$$x + \frac{b}{2a} = \pm\frac{\sqrt{b^2 - 4ac}}{2a}$$
$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

**Example 1.** Solve $3x^2 + 4x - 2 = 0$.
$a = 3,\ b = 4,\ c = -2$; $b^2 - 4ac = 16 + 24 = 40$.
$$x = \frac{-4 \pm \sqrt{40}}{6} = \frac{-4 \pm 2\sqrt{10}}{6} = \frac{-2 \pm \sqrt{10}}{3}$$

**Example 2 (complex roots).** Solve $x^2 + 2x + 5 = 0$.
$b^2 - 4ac = 4 - 20 = -16$.
$$x = \frac{-2 \pm \sqrt{-16}}{2} = \frac{-2 \pm 4i}{2} = -1 \pm 2i$$

### 3.4 Graphical Method

Plot $y = ax^2 + bx + c$; the real roots are the $x$-coordinates where the curve meets the $x$-axis. Useful for estimating roots and visualising how many real roots exist, but not exact.

### 3.5 Which Method to Use?

| Situation | Best method |
|-----------|-------------|
| Small integer coefficients, "nice" roots | Factorisation |
| $b = 0$ or $c = 0$ | Direct (square root / take out common factor) |
| Need vertex form or a derivation | Completing the square |
| Anything else / roots irrational or complex | Quadratic formula (always works) |

````python
import cmath

def solve_quadratic(a, b, c):
    """Return the two roots of ax^2 + bx + c = 0 (complex if D < 0)."""
    if a == 0:
        raise ValueError("a must be non-zero for a quadratic")
    D = b**2 - 4*a*c
    sqrt_D = cmath.sqrt(D)
    r1 = (-b + sqrt_D) / (2*a)
    r2 = (-b - sqrt_D) / (2*a)
    # Show real roots as plain floats
    if D >= 0:
        r1, r2 = r1.real, r2.real
    return D, r1, r2

for a, b, c in [(6, -5, -6), (2, -8, 3), (3, 4, -2), (1, 2, 5)]:
    D, r1, r2 = solve_quadratic(a, b, c)
    print(f"{a}x^2 + ({b})x + ({c}) = 0  ->  D = {D:>4},  roots = {r1:.4f}, {r2:.4f}")
````

## 4. The Discriminant and Nature of Roots

The **discriminant** is

$$D = b^2 - 4ac$$

It is the quantity under the square root in the formula, so it decides what kind of roots we get.

| Value of $D$ | Nature of roots | Graph of $y = ax^2+bx+c$ |
|--------------|-----------------|--------------------------|
| $D > 0$ | Two distinct real roots | Cuts $x$-axis at two points |
| $D = 0$ | Two equal real roots, $x = -\tfrac{b}{2a}$ | Touches $x$-axis at the vertex |
| $D < 0$ | No real roots; two complex conjugate roots $p \pm qi$ | Does not meet $x$-axis |

**Finer classification when $a, b, c$ are rational:**
- $D > 0$ and a **perfect square** → roots are real, distinct and **rational**.
- $D > 0$ and **not** a perfect square → roots are real, distinct and **irrational**, occurring as a conjugate pair $p \pm \sqrt{q}$.
- $D = 0$ → roots are real, equal and rational.

**Other useful facts:**
- If $a$ and $c$ have opposite signs ($ac < 0$), then $D = b^2 - 4ac > 0$ automatically → two real roots (of opposite signs).
- Complex roots (with real coefficients) always come in conjugate pairs: if $p + qi$ is a root, so is $p - qi$.
- Difference of roots: $|\alpha - \beta| = \dfrac{\sqrt{D}}{|a|}$.

**Example.** For what $k$ does $x^2 + kx + 9 = 0$ have equal roots?
$D = k^2 - 36 = 0 \Rightarrow k = \pm 6$.

**Example.** For what $k$ does $kx^2 + 4x + 1 = 0$ have real roots?
Need $D = 16 - 4k \geq 0 \Rightarrow k \leq 4$, and $k \neq 0$ for the equation to be quadratic. So $k \leq 4,\ k \neq 0$.

## 5. Relation Between Roots and Coefficients (Vieta's Formulas)

If $\alpha, \beta$ are the roots of $ax^2 + bx + c = 0$:

$$\boxed{\alpha + \beta = -\frac{b}{a}} \qquad \boxed{\alpha\beta = \frac{c}{a}}$$

*Proof:* $ax^2 + bx + c = a(x - \alpha)(x - \beta) = a x^2 - a(\alpha + \beta)x + a\alpha\beta$. Compare coefficients.

### Symmetric expressions in terms of $S = \alpha + \beta$ and $P = \alpha\beta$

| Expression | In terms of $S$ and $P$ |
|------------|-------------------------|
| $\alpha^2 + \beta^2$ | $S^2 - 2P$ |
| $(\alpha - \beta)^2$ | $S^2 - 4P$ |
| $\lvert\alpha - \beta\rvert$ | $\sqrt{S^2 - 4P}$ |
| $\alpha^2 - \beta^2$ | $S(\alpha - \beta)$ |
| $\alpha^3 + \beta^3$ | $S^3 - 3PS$ |
| $\alpha^3 - \beta^3$ | $(\alpha - \beta)(S^2 - P)$ |
| $\alpha^4 + \beta^4$ | $(S^2 - 2P)^2 - 2P^2$ |
| $\dfrac{1}{\alpha} + \dfrac{1}{\beta}$ | $\dfrac{S}{P}$ |
| $\dfrac{1}{\alpha^2} + \dfrac{1}{\beta^2}$ | $\dfrac{S^2 - 2P}{P^2}$ |
| $\dfrac{\alpha}{\beta} + \dfrac{\beta}{\alpha}$ | $\dfrac{S^2 - 2P}{P}$ |
| $\alpha^2\beta + \alpha\beta^2$ | $PS$ |

**Example.** If $\alpha, \beta$ are roots of $2x^2 - 3x - 5 = 0$, find $\alpha^2 + \beta^2$ and $\tfrac{1}{\alpha} + \tfrac{1}{\beta}$.

$S = \tfrac{3}{2},\ P = -\tfrac{5}{2}$.
- $\alpha^2 + \beta^2 = \tfrac{9}{4} + 5 = \tfrac{29}{4}$
- $\tfrac{1}{\alpha} + \tfrac{1}{\beta} = \dfrac{3/2}{-5/2} = -\tfrac{3}{5}$

(Check: the roots are $\tfrac{5}{2}$ and $-1$; $\tfrac{25}{4} + 1 = \tfrac{29}{4}$. ✓)

### Signs of the roots (real roots, $D \geq 0$)

| Condition | Roots |
|-----------|-------|
| $P > 0,\ S > 0$ | both positive |
| $P > 0,\ S < 0$ | both negative |
| $P < 0$ | opposite signs |
| $P < 0,\ S > 0$ | opposite signs, positive root has larger magnitude |
| $P < 0,\ S < 0$ | opposite signs, negative root has larger magnitude |
| $S = 0$ ($b = 0$) | equal in magnitude, opposite in sign |
| $P = 1$ ($c = a$) | reciprocals of each other |
| $P = 0$ ($c = 0$) | one root is 0 |

````python
# Verify Vieta's formulas on 2x^2 - 3x - 5 = 0
a, b, c = 2, -3, -5
_, alpha, beta = solve_quadratic(a, b, c)
print("alpha + beta =", alpha + beta, "  -b/a =", -b/a)
print("alpha * beta =", alpha * beta, "  c/a =", c/a)
print("alpha^2 + beta^2 =", alpha**2 + beta**2, "  (S^2 - 2P) =", (-b/a)**2 - 2*(c/a))
````

## 6. Forming a Quadratic from Its Roots

If the roots are $\alpha$ and $\beta$, the equation (with leading coefficient 1) is

$$\boxed{x^2 - (\alpha + \beta)x + \alpha\beta = 0}$$

i.e. $x^2 - (\text{sum})x + (\text{product}) = 0$.

**Example.** Roots $3$ and $-5$: sum $= -2$, product $= -15$ → $x^2 + 2x - 15 = 0$.

**Example.** Roots $2 + \sqrt{3}$ and $2 - \sqrt{3}$: sum $= 4$, product $= 4 - 3 = 1$ → $x^2 - 4x + 1 = 0$.

### Transformed roots

Given $\alpha, \beta$ are roots of $ax^2 + bx + c = 0$, the equation whose roots are:

| New roots | New equation | Rule |
|-----------|--------------|------|
| $\alpha + k,\ \beta + k$ | $a(x-k)^2 + b(x-k) + c = 0$ | replace $x$ by $x - k$ |
| $k\alpha,\ k\beta$ | $ax^2 + bkx + ck^2 = 0$ | replace $x$ by $x/k$ |
| $-\alpha,\ -\beta$ | $ax^2 - bx + c = 0$ | replace $x$ by $-x$ |
| $\tfrac{1}{\alpha},\ \tfrac{1}{\beta}$ | $cx^2 + bx + a = 0$ | replace $x$ by $1/x$ (reverse coefficients) |
| $\alpha^2,\ \beta^2$ | use new sum $S^2 - 2P$ and product $P^2$ | |

**Example.** $\alpha, \beta$ are roots of $x^2 - 3x + 1 = 0$. Find the equation with roots $\alpha^2, \beta^2$.
New sum $= 9 - 2 = 7$, new product $= 1$ → $x^2 - 7x + 1 = 0$.

## 7. Graph of a Quadratic Function

The graph of $y = f(x) = ax^2 + bx + c$ is a **parabola**.

### Vertex form

Completing the square gives

$$y = a\left(x + \frac{b}{2a}\right)^2 - \frac{D}{4a}$$

or, in general, $y = a(x - h)^2 + k$.

| Feature | Value |
|---------|-------|
| Vertex | $\left(-\dfrac{b}{2a},\ -\dfrac{D}{4a}\right)$ |
| Axis of symmetry | $x = -\dfrac{b}{2a}$ |
| $y$-intercept | $(0, c)$ |
| $x$-intercepts | real roots (if $D \geq 0$) |
| Opens upward | $a > 0$ (vertex is a minimum) |
| Opens downward | $a < 0$ (vertex is a maximum) |

The axis of symmetry lies exactly midway between the roots: $-\tfrac{b}{2a} = \tfrac{\alpha + \beta}{2}$.

### The six basic shapes

```
 a > 0, D > 0        a > 0, D = 0        a > 0, D < 0
   \       /           \     /             \     /
----\-----/----         \   /               \   /
     \___/          -----\_/-----            \_/
                                       ---------------
 two real roots      one repeated root   no real roots

 a < 0, D > 0        a < 0, D = 0        a < 0, D < 0
      ___                                ---------------
     /   \          -----/‾\-----            /‾\
----/-----\----         /   \               /   \
   /       \           /     \             /     \
```

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 400)
cases = [
    ("a > 0, D > 0", 1, 0, -2), ("a > 0, D = 0", 1, 0, 0), ("a > 0, D < 0", 1, 0, 2),
    ("a < 0, D > 0", -1, 0, 2), ("a < 0, D = 0", -1, 0, 0), ("a < 0, D < 0", -1, 0, -2),
]
fig, axes = plt.subplots(2, 3, figsize=(12, 7), sharex=True, sharey=True)
for ax, (title, a, b, c) in zip(axes.flat, cases):
    ax.plot(x, a*x**2 + b*x + c)
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    sign = "-" if a < 0 else ""
    const = f" + {c}" if c > 0 else (f" - {-c}" if c < 0 else "")
    ax.set_title(f"{title}\n$y = {sign}x^2{const}$")
    ax.set_ylim(-6, 6)
    ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

**Example.** Sketch $y = 2x^2 - 8x + 3$.
- $a = 2 > 0$, opens upward.
- Vertex: $x = \tfrac{8}{4} = 2$, $y = 8 - 16 + 3 = -5$ → $(2, -5)$.
  (Check: $D = 64 - 24 = 40$, $-\tfrac{D}{4a} = -\tfrac{40}{8} = -5$ ✓)
- $y$-intercept: $(0, 3)$.
- Roots: $2 \pm \tfrac{\sqrt{10}}{2} \approx 0.42,\ 3.58$.

````python
# Worked example: y = 2x^2 - 8x + 3 with its vertex and roots
a, b, c = 2, -8, 3
D, r1, r2 = solve_quadratic(a, b, c)
xv, yv = -b/(2*a), -D/(4*a)

x = np.linspace(-1, 5, 400)
plt.figure(figsize=(6, 4))
plt.plot(x, a*x**2 + b*x + c, label="$y = 2x^2 - 8x + 3$")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(xv, linestyle="--", color="gray", label=f"axis $x = {xv:g}$")
plt.scatter([xv], [yv], color="red", zorder=3, label=f"vertex ({xv:g}, {yv:g})")
plt.scatter([r1, r2], [0, 0], color="green", zorder=3, label=f"roots {r2:.2f}, {r1:.2f}")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

## 8. Sign of a Quadratic Expression and Quadratic Inequalities

For $f(x) = ax^2 + bx + c$:

| Case | Sign of $f(x)$ |
|------|----------------|
| $D < 0$ | $f(x)$ has the **same sign as $a$** for every real $x$ |
| $D = 0$ | same sign as $a$ for all $x$ except $x = -\tfrac{b}{2a}$, where $f = 0$ |
| $D > 0$, roots $\alpha < \beta$ | same sign as $a$ **outside** $[\alpha, \beta]$; opposite sign to $a$ **between** $\alpha$ and $\beta$ |

Consequences:
- $ax^2 + bx + c > 0$ for **all** real $x$ $\iff a > 0$ and $D < 0$.
- $ax^2 + bx + c < 0$ for **all** real $x$ $\iff a < 0$ and $D < 0$.

### Solving quadratic inequalities

1. Bring everything to one side: $f(x) > 0$ (or $<, \geq, \leq$).
2. Make $a > 0$ (multiply by $-1$ and flip the inequality if needed).
3. Find the roots $\alpha < \beta$.
4. For $a > 0$: $f(x) > 0 \iff x < \alpha$ or $x > \beta$; $f(x) < 0 \iff \alpha < x < \beta$.

**Example.** $x^2 - 5x + 6 < 0$ → $(x-2)(x-3) < 0$ → $2 < x < 3$.

**Example.** $x^2 - x - 12 > 0$ → $(x-4)(x+3) > 0$ → $x < -3$ or $x > 4$.

**Example.** $-x^2 + 4x - 3 \geq 0$ → $x^2 - 4x + 3 \leq 0$ → $(x-1)(x-3) \leq 0$ → $1 \leq x \leq 3$.

## 9. Maximum and Minimum Values

From the vertex form $f(x) = a\left(x + \frac{b}{2a}\right)^2 - \frac{D}{4a}$:

- If $a > 0$: **minimum** value $= -\dfrac{D}{4a} = \dfrac{4ac - b^2}{4a}$ at $x = -\dfrac{b}{2a}$; no maximum.
- If $a < 0$: **maximum** value $= -\dfrac{D}{4a} = \dfrac{4ac - b^2}{4a}$ at $x = -\dfrac{b}{2a}$; no minimum.

**Example.** Minimum of $x^2 - 6x + 11$: $(x - 3)^2 + 2$ → minimum $2$ at $x = 3$.

**Example (projectile).** A ball's height is $h(t) = -5t^2 + 20t + 1$ metres. Max height occurs at $t = -\tfrac{20}{2(-5)} = 2$ s, and $h(2) = -20 + 40 + 1 = 21$ m.

**Example (fixed perimeter).** A rectangle has perimeter 40. Its area is $A = x(20 - x) = -x^2 + 20x$, maximised at $x = 10$ → a $10 \times 10$ square, area 100.

**Range of $f(x) = ax^2 + bx + c$ over all reals:**
- $a > 0$: $\left[-\tfrac{D}{4a}, \infty\right)$
- $a < 0$: $\left(-\infty, -\tfrac{D}{4a}\right]$

## 10. Common Roots

Consider
$$a_1x^2 + b_1x + c_1 = 0, \qquad a_2x^2 + b_2x + c_2 = 0.$$

**Both roots common:**
$$\frac{a_1}{a_2} = \frac{b_1}{b_2} = \frac{c_1}{c_2}$$

**Exactly one root common** (condition):
$$(c_1a_2 - c_2a_1)^2 = (b_1c_2 - b_2c_1)(a_1b_2 - a_2b_1)$$

**Practical method:** subtract suitable multiples of the equations to eliminate $x^2$, solve for the common root, then substitute back. Or, if one equation factors easily, try each of its roots in the other.

**Example.** For what $k$ do $x^2 - 3x + 2 = 0$ and $x^2 - kx + 6 = 0$ share a root?
The first has roots $1$ and $2$.
- Common root $1$: $1 - k + 6 = 0 \Rightarrow k = 7$.
- Common root $2$: $4 - 2k + 6 = 0 \Rightarrow k = 5$.

So $k = 5$ or $k = 7$.

## 11. Equations Reducible to Quadratic Form

The trick is always a **substitution** that turns the equation into $at^2 + bt + c = 0$.

### 11.1 Biquadratic: $ax^4 + bx^2 + c = 0$
Let $t = x^2$.

**Example.** $x^4 - 5x^2 + 4 = 0$ → $t^2 - 5t + 4 = 0$ → $t = 1, 4$ → $x = \pm 1, \pm 2$.

*Remember: $t = x^2$ must be $\geq 0$ for real $x$; reject negative $t$.*

### 11.2 Repeated expression: $a[f(x)]^2 + b f(x) + c = 0$
Let $t = f(x)$.

**Example.** $(x^2 + x)^2 - 8(x^2 + x) + 12 = 0$.
$t = x^2 + x$: $t^2 - 8t + 12 = 0 \Rightarrow t = 2$ or $6$.
- $x^2 + x - 2 = 0 \Rightarrow x = 1, -2$
- $x^2 + x - 6 = 0 \Rightarrow x = 2, -3$

### 11.3 Reciprocal type: $x + \tfrac{1}{x} = k$
Multiply through by $x$.

**Example.** $x + \tfrac{1}{x} = \tfrac{5}{2}$ → $2x^2 - 5x + 2 = 0$ → $(2x - 1)(x - 2) = 0$ → $x = 2, \tfrac{1}{2}$.

### 11.4 Exponential type
**Example.** $4^x - 6 \cdot 2^x + 8 = 0$. Let $t = 2^x > 0$: $t^2 - 6t + 8 = 0 \Rightarrow t = 2, 4 \Rightarrow x = 1, 2$.

### 11.5 Radical equations
Isolate the radical, square both sides, solve — then **check every answer in the original equation**, since squaring can introduce **extraneous roots**.

**Example.** $\sqrt{x + 7} = x + 1$.
Square: $x + 7 = x^2 + 2x + 1 \Rightarrow x^2 + x - 6 = 0 \Rightarrow x = 2$ or $x = -3$.
Check $x = -3$: $\sqrt{4} = 2$ but $x + 1 = -2$. ✗ Rejected.
Check $x = 2$: $\sqrt{9} = 3 = 2 + 1$. ✓
**Answer:** $x = 2$.

### 11.6 Rational (fractional) equations
Multiply by the LCD, solve, and reject any value that makes a denominator zero.

**Example.** $\dfrac{1}{x} + \dfrac{1}{x - 2} = \dfrac{4}{3}$.
LCD $3x(x-2)$: $3(x - 2) + 3x = 4x(x - 2)$ → $6x - 6 = 4x^2 - 8x$ → $4x^2 - 14x + 6 = 0$ → $2x^2 - 7x + 3 = 0$ → $(2x - 1)(x - 3) = 0$ → $x = \tfrac{1}{2}, 3$. Neither is $0$ or $2$, so both are valid.

## 12. Location of Roots (Advanced)

Let $f(x) = ax^2 + bx + c$ with real roots $\alpha \leq \beta$, and let $k, k_1 < k_2$ be real numbers. Write $x_v = -\tfrac{b}{2a}$ for the vertex.

| Requirement | Conditions |
|-------------|-----------|
| Both roots greater than $k$ | $D \geq 0,\ \ a\,f(k) > 0,\ \ x_v > k$ |
| Both roots less than $k$ | $D \geq 0,\ \ a\,f(k) > 0,\ \ x_v < k$ |
| $k$ lies between the roots | $a\,f(k) < 0$ (this alone suffices) |
| Both roots in $(k_1, k_2)$ | $D \geq 0,\ \ a\,f(k_1) > 0,\ \ a\,f(k_2) > 0,\ \ k_1 < x_v < k_2$ |
| Exactly one root in $(k_1, k_2)$ | $f(k_1)\,f(k_2) < 0$ |

**Example.** Find $m$ so that both roots of $x^2 - (m - 3)x + m = 0$ are positive.
- $D \geq 0$: $(m - 3)^2 - 4m \geq 0 \Rightarrow m^2 - 10m + 9 \geq 0 \Rightarrow m \leq 1$ or $m \geq 9$.
- Sum $> 0$: $m - 3 > 0 \Rightarrow m > 3$.
- Product $> 0$: $m > 0$.

Intersection: $m \geq 9$.

## 13. Word Problems

**General strategy**
1. Define the unknown clearly (with units).
2. Translate the statement into an equation.
3. Reduce to $ax^2 + bx + c = 0$ and solve.
4. **Reject roots that don't make sense** (negative lengths, negative speeds, non-integers when integers are required, etc.).
5. Answer the question that was actually asked.

**Example 1 (numbers).** The product of two consecutive positive integers is 132. Find them.
$n(n + 1) = 132 \Rightarrow n^2 + n - 132 = 0 \Rightarrow (n + 12)(n - 11) = 0$.
$n = 11$ (reject $-12$). **Answer:** 11 and 12.

**Example 2 (speed–time).** A train covers 360 km at a uniform speed. Had the speed been 5 km/h more, it would have taken 1 hour less. Find the speed.
$$\frac{360}{v} - \frac{360}{v + 5} = 1 \Rightarrow 360 \cdot 5 = v(v + 5) \Rightarrow v^2 + 5v - 1800 = 0$$
$(v + 45)(v - 40) = 0 \Rightarrow v = 40$ km/h.

**Example 3 (geometry).** A rectangle has perimeter 28 m and area 40 m². Find its sides.
$l + w = 14,\ lw = 40$ → the sides are roots of $x^2 - 14x + 40 = 0 = (x - 4)(x - 10)$.
**Answer:** 10 m × 4 m.

**Example 4 (ages).** The product of a person's age 5 years ago and their age 5 years from now is 11. Find their present age.
$(x - 5)(x + 5) = 11 \Rightarrow x^2 = 36 \Rightarrow x = 6$ (reject $-6$). **Answer:** 6 years.

**Common word-problem templates**

| Type | Typical equation |
|------|------------------|
| Consecutive integers | $n(n+1) = P$ |
| Number + reciprocal | $x + \tfrac{1}{x} = k$ |
| Speed / time / distance | $\tfrac{d}{v} - \tfrac{d}{v + u} = \Delta t$ |
| Upstream / downstream | speeds $v - u$ and $v + u$ |
| Work done | $\tfrac{1}{x} + \tfrac{1}{x + k} = \tfrac{1}{T}$ |
| Area / perimeter | $l + w = \tfrac{p}{2},\ lw = A$ |
| Right triangle | Pythagoras: $x^2 + (x + k)^2 = h^2$ |
| Projectile height | $h(t) = -\tfrac{1}{2}gt^2 + v_0t + h_0$ |

## 14. Quick Tricks and Common Mistakes

### Tricks
- **If $a + b + c = 0$**, then $x = 1$ is a root; the other root is $\tfrac{c}{a}$.
  e.g. $3x^2 - 7x + 4 = 0$ → roots $1,\ \tfrac{4}{3}$.
- **If $a - b + c = 0$**, then $x = -1$ is a root; the other root is $-\tfrac{c}{a}$.
  e.g. $2x^2 + 5x + 3 = 0$ → roots $-1,\ -\tfrac{3}{2}$.
- **Even $b$:** if $b = 2b'$, then $x = \dfrac{-b' \pm \sqrt{b'^2 - ac}}{a}$ (less arithmetic).
- **Checking answers:** verify sum $= -b/a$ and product $= c/a$ — much faster than substituting.
- If one root is known, the other is $\tfrac{c}{a\alpha}$ or $-\tfrac{b}{a} - \alpha$.

### Common mistakes
| Mistake | Correct approach |
|---------|------------------|
| Dividing both sides by $x$ (e.g. $x^2 = 3x \Rightarrow x = 3$) | Factor: $x(x - 3) = 0 \Rightarrow x = 0, 3$ |
| Forgetting $\pm$ when taking square roots | $x^2 = 9 \Rightarrow x = \pm 3$ |
| Sign error on $b$: using $b$ instead of $-b$ | Formula starts with $-b$ |
| Dividing only $\sqrt{D}$ by $2a$ | The whole numerator $-b \pm \sqrt{D}$ is over $2a$ |
| Treating $-b^2$ as $(-b)^2$ | $D = b^2 - 4ac$ — square $b$ first, so $b^2 \geq 0$ always |
| Not checking radical / rational equations | Substitute back; reject extraneous roots |
| Forgetting the condition $a \neq 0$ in "find $k$" problems | State it explicitly |
| Flipping inequality wrongly | Only flip when multiplying/dividing by a negative |
| Writing $\alpha + \beta = b/a$ | It is $-b/a$ |
