## 1. Practice with the Expansion Formulae

> The lecturer's routine for each question: **pause, attempt it yourself, then compare** with the solution.

### Q1. Prove $\sin(A+B)\,\sin(A-B) = \sin^2 A - \sin^2 B$
Expand both factors:
$$\text{LHS} = (\sin A\cos B + \cos A\sin B)(\sin A\cos B - \cos A\sin B)$$
This is $(x + y)(x - y) = x^2 - y^2$:
$$= \sin^2A\cos^2B - \cos^2A\sin^2B$$

**Plan:** the target contains only $\sin$, so convert every $\cos^2$ into $1 - \sin^2$:
$$= \sin^2A(1 - \sin^2B) - (1 - \sin^2A)\sin^2B$$
$$= \sin^2A - \sin^2A\sin^2B - \sin^2B + \sin^2A\sin^2B = \sin^2A - \sin^2B \quad\blacksquare$$
(the two $\sin^2A\sin^2B$ terms cancel)

> **JEE tip:** memorise this result as a formula, so you can use it directly.

### Q2. Prove $\cos(A+B)\,\cos(A-B) = \cos^2 A - \sin^2 B$
$$\text{LHS} = (\cos A\cos B - \sin A\sin B)(\cos A\cos B + \sin A\sin B) = \cos^2A\cos^2B - \sin^2A\sin^2B$$

**Look at the target:** it has $\cos$ with $A$ and $\sin$ with $B$. So keep the $A$-terms as $\cos$ and the $B$-terms as $\sin$:
$$= \cos^2A(1 - \sin^2B) - (1 - \cos^2A)\sin^2B$$
$$= \cos^2A - \cos^2A\sin^2B - \sin^2B + \cos^2A\sin^2B = \cos^2A - \sin^2B \quad\blacksquare$$
(the two $\cos^2A\sin^2B$ terms cancel)

> **Lesson:** decide your destination first. The form of the right-hand side tells you which conversion to make.

```python
A, B = rng.uniform(-np.pi, np.pi, 100_000), rng.uniform(-np.pi, np.pi, 100_000)
verify("Q1  sin(A+B)sin(A−B) = sin²A − sin²B", np.sin(A + B)*np.sin(A - B), np.sin(A)**2 - np.sin(B)**2)
verify("Q2  cos(A+B)cos(A−B) = cos²A − sin²B", np.cos(A + B)*np.cos(A - B), np.cos(A)**2 - np.sin(B)**2)

# Visual: fix B and sweep A
Ad = np.linspace(-180, 180, 721); Ar = d(Ad); Bv = d(40)
fig, axes = plt.subplots(1, 2, figsize=(12, 3.6), sharey=True)
for ax, lhs, rhs, title in [
        (axes[0], np.sin(Ar + Bv)*np.sin(Ar - Bv), np.sin(Ar)**2 - np.sin(Bv)**2, "Q1: sin(A+B)·sin(A−B)  vs  sin²A − sin²B"),
        (axes[1], np.cos(Ar + Bv)*np.cos(Ar - Bv), np.cos(Ar)**2 - np.sin(Bv)**2, "Q2: cos(A+B)·cos(A−B)  vs  cos²A − sin²B")]:
    ax.plot(Ad, lhs, color=BLUE, lw=5, alpha=0.4, label="product (LHS)")
    ax.plot(Ad, rhs, color=ORANGE, lw=1.6, ls="--", label="difference of squares (RHS)")
    ax.axhline(0, color=INK2, lw=1); ax.set_xticks(range(-180, 181, 90)); ax.set_xlim(-180, 180)
    ax.set_xlabel("A (degrees), with B = 40°"); ax.set_title(title, fontsize=11)
axes[0].legend(frameon=False, fontsize=9, loc="lower center")
plt.tight_layout(); plt.show()
```

### Q3. Prove $\sin(45^\circ + A)\cos(45^\circ - B) + \cos(45^\circ + A)\sin(45^\circ - B) = \cos(A - B)$

**Spot the pattern.** The LHS has the shape $\sin x\cos y + \cos x\sin y$, which is the expansion of $\sin(x + y)$, with $x = 45^\circ + A$ and $y = 45^\circ - B$:
$$\text{LHS} = \sin\big[(45^\circ + A) + (45^\circ - B)\big] = \sin\big(90^\circ + (A - B)\big)$$

Allied angles: $90^\circ + \theta$ is in quadrant II, where $\sin > 0$, and it is a y-axis angle, so $\sin \to \cos$:
$$= \cos(A - B) = \text{RHS} \quad\blacksquare$$

> **Lesson:** the formulas work in **both directions**. You can recognise an expanded form and collapse it back.

```python
A, B = rng.uniform(-np.pi, np.pi, 100_000), rng.uniform(-np.pi, np.pi, 100_000)
q = np.pi/4
verify("Q3", np.sin(q + A)*np.cos(q - B) + np.cos(q + A)*np.sin(q - B), np.cos(A - B))
```

### Q4. Prove $\dfrac{\sin(A-B)}{\cos A\cos B} + \dfrac{\sin(B-C)}{\cos B\cos C} + \dfrac{\sin(C-A)}{\cos C\cos A} = 0$

Take the first term and expand the numerator. "**Everyone has a right to the denominator**", so split the fraction:
$$\frac{\sin(A-B)}{\cos A\cos B} = \frac{\sin A\cos B - \cos A\sin B}{\cos A\cos B} = \frac{\sin A\cos B}{\cos A\cos B} - \frac{\cos A\sin B}{\cos A\cos B} = \tan A - \tan B$$

By the same steps:
$$\frac{\sin(B-C)}{\cos B\cos C} = \tan B - \tan C, \qquad \frac{\sin(C-A)}{\cos C\cos A} = \tan C - \tan A$$

Adding them, everything cancels:
$$(\tan A - \tan B) + (\tan B - \tan C) + (\tan C - \tan A) = 0 \quad\blacksquare$$

It is like walking from $\tan A$ to $\tan B$ to $\tan C$ and back to $\tan A$: you end where you started, so the total displacement is $0$.

```python
A, B, C = rng.uniform(-1.4, 1.4, (3, 100_000))
T = lambda X, Y: np.sin(X - Y)/(np.cos(X)*np.cos(Y))
verify("Q4  sum of three terms = 0", T(A, B) + T(B, C) + T(C, A), np.zeros_like(A))

# Visual: each term is a jump between tan values on a number line
Av, Bv, Cv = d(55), d(20), d(-35)
tA, tB, tC = np.tan([Av, Bv, Cv])
fig, ax = plt.subplots(figsize=(10, 2.8)); ax.grid(False)
ax.axhline(0, color=INK2, lw=1); ax.set_yticks([])
for v, lab in [(tA, "tan A"), (tB, "tan B"), (tC, "tan C")]:
    ax.plot(v, 0, "o", color=INK, ms=8, mec="white", mew=1.5, zorder=5)
    ax.text(v, -0.25, f"{lab}\n{v:.3f}", ha="center", va="top", fontsize=9)
for (x0, x1, h, col, lab) in [(tA, tB, 1.75, BLUE, "sin(A−B)/(cosA cosB) = tanA − tanB"),
                              (tB, tC, 1.5, ORANGE, "sin(B−C)/(cosB cosC) = tanB − tanC"),
                              (tC, tA, 1.25, AQUA, "sin(C−A)/(cosC cosA) = tanC − tanA")]:
    ax.add_patch(FancyArrowPatch((x0, 0.05), (x1, 0.05), connectionstyle=f"arc3,rad={0.3 if x1 < x0 else -0.3}",
                                 arrowstyle="-|>", mutation_scale=14, color=col, lw=1.8))
    ax.text(min(tA, tB, tC) - 0.35, h, "→ " + lab, ha="left", fontsize=9, color=col)
ax.set_ylim(-0.9, 2.0); ax.set_xlim(min(tA, tB, tC) - 0.4, max(tA, tB, tC) + 0.4)
ax.set_title("Q4: the three terms are jumps that bring you back to the start, so the total is 0", fontsize=11)
plt.tight_layout(); plt.show()
```

### Q5. If $A + B + C = \pi$, prove $\tan A + \tan B + \tan C = \tan A\,\tan B\,\tan C$
(This appeared in an IIT-JEE paper in 1978 and is now a solved example in NCERT Class 11.)

From $A + B + C = \pi$:
$$A + B = \pi - C$$
**If two angles are equal, their tangents are equal**, so take $\tan$ of both sides:
$$\tan(A + B) = \tan(\pi - C)$$

- **LHS:** expansion formula, $\dfrac{\tan A + \tan B}{1 - \tan A\tan B}$
- **RHS:** $\pi - C$ is in quadrant II, where $\tan < 0$, and $\pi$ is on the x-axis, so $\tan$ stays $\tan$. That gives $-\tan C$.

$$\frac{\tan A + \tan B}{1 - \tan A\tan B} = -\tan C$$

Cross-multiply:
$$\tan A + \tan B = -\tan C + \tan A\tan B\tan C$$
$$\boxed{\tan A + \tan B + \tan C = \tan A\tan B\tan C} \quad\blacksquare$$

**Interpretation:** in **any triangle** $ABC$ (where $A + B + C = \pi$), the **sum** of the tangents equals their **product**. That is the "add or multiply, get the same answer" idea from Lecture 1, now for tangents.

> **JEE / advanced tip:** memorise it as: *in $\triangle ABC$, $\sum\tan A = \prod\tan A$*. The converse is also true.

```python
# Random triangles: pick A, B, set C = π − A − B (skip right angles, where tan is undefined)
n = 4000
A = rng.uniform(0.05, np.pi - 0.1, n); B = rng.uniform(0.05, np.pi - A - 0.05)
C = np.pi - A - B
keep = (np.abs(np.cos(A)) > 0.05) & (np.abs(np.cos(B)) > 0.05) & (np.abs(np.cos(C)) > 0.05)
A, B, C = A[keep], B[keep], C[keep]
s = np.tan(A) + np.tan(B) + np.tan(C); p = np.tan(A)*np.tan(B)*np.tan(C)
verify("Q5  tanA + tanB + tanC = tanA·tanB·tanC  (triangles)", s, p)

fig, ax = plt.subplots(figsize=(6.2, 5.6))
lim = 12
m = (np.abs(s) < lim) & (np.abs(p) < lim)
obtuse = np.max(np.c_[A, B, C], axis=1) > np.pi/2
ax.scatter(s[m & ~obtuse], p[m & ~obtuse], s=10, color=BLUE, alpha=0.6, label="acute triangles")
ax.scatter(s[m & obtuse], p[m & obtuse], s=10, color=ORANGE, alpha=0.6, label="obtuse triangles")
ax.plot([-lim, lim], [-lim, lim], color=INK2, lw=1, ls="--")
ax.text(lim*0.4, lim*0.62, "sum = product", color=INK2, rotation=45, fontsize=9)
ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim); ax.set_aspect("equal")
ax.set_xlabel("tan A + tan B + tan C"); ax.set_ylabel("tan A · tan B · tan C")
ax.legend(frameon=False, loc="upper left", fontsize=9)
ax.set_title("Every random triangle lands exactly on the line y = x", fontsize=11)
plt.tight_layout(); plt.show()
```

### Q6. Prove $\tan 3x\,\tan 2x\,\tan x = \tan 3x - \tan 2x - \tan x$
**Hint from the lecture:** it is solved exactly like Q5.

Everyone knows $x + 2x = 3x$. Equal angles have equal tangents:
$$\tan(x + 2x) = \tan 3x \;\Longrightarrow\; \frac{\tan x + \tan 2x}{1 - \tan x\tan 2x} = \tan 3x$$

Cross-multiply:
$$\tan x + \tan 2x = \tan 3x - \tan x\tan 2x\tan 3x$$

Rearrange, moving the product to the left and the single terms to the right:
$$\boxed{\tan 3x\tan 2x\tan x = \tan 3x - \tan 2x - \tan x} \quad\blacksquare$$

> **Advanced tip (optional):** the general version is $\tan(A+B) = \tan C$, which gives $\tan A\tan B\tan C = \tan C - \tan A - \tan B$ whenever $A + B = C$.

```python
x = np.linspace(-np.pi/2 + 0.01, np.pi/2 - 0.01, 6000)
lhs = np.tan(3*x)*np.tan(2*x)*np.tan(x); rhs = np.tan(3*x) - np.tan(2*x) - np.tan(x)
verify("Q6", lhs, rhs)

def broken(y, jump=8):
    y = y.copy(); y[np.abs(np.diff(np.r_[y[0], y])) > jump] = np.nan; return y
fig, ax = plt.subplots(figsize=(10, 3.8))
ax.plot(np.rad2deg(x), broken(lhs), color=BLUE, lw=5, alpha=0.4, label="tan3x · tan2x · tanx")
ax.plot(np.rad2deg(x), broken(rhs), color=ORANGE, lw=1.5, ls="--", label="tan3x − tan2x − tanx")
ax.axhline(0, color=INK2, lw=1)
ax.set_ylim(-10, 10); ax.set_xlim(-90, 90); ax.set_xticks(range(-90, 91, 30)); ax.set_xlabel("x (degrees)")
ax.legend(frameon=True, framealpha=1, edgecolor="none", fontsize=9, loc="upper left")
ax.set_title("Q6: both sides coincide everywhere they are defined", fontsize=11)
plt.tight_layout(); plt.show()
```

## 2. Product-to-Sum Formulae

Write down the four expansion formulae:
$$\sin(A+B) = \sin A\cos B + \cos A\sin B \qquad (1)$$
$$\sin(A-B) = \sin A\cos B - \cos A\sin B \qquad (2)$$
$$\cos(A+B) = \cos A\cos B - \sin A\sin B \qquad (3)$$
$$\cos(A-B) = \cos A\cos B + \sin A\sin B \qquad (4)$$

**Add and subtract them in pairs:**
- $(1) + (2)$: the $\cos A\sin B$ terms cancel, giving $2\sin A\cos B$
- $(1) - (2)$: the $\sin A\cos B$ terms cancel, giving $2\cos A\sin B$
- $(4) + (3)$: the $\sin A\sin B$ terms cancel, giving $2\cos A\cos B$
- $(4) - (3)$: the $\cos A\cos B$ terms cancel, giving $2\sin A\sin B$. Note the order: $\cos(A-B)$ **minus** $\cos(A+B)$.

$$\boxed{2\sin A\cos B = \sin(A+B) + \sin(A-B)}$$
$$\boxed{2\cos A\sin B = \sin(A+B) - \sin(A-B)}$$
$$\boxed{2\cos A\cos B = \cos(A+B) + \cos(A-B)}$$
$$\boxed{2\sin A\sin B = \cos(A-B) - \cos(A+B)}$$

**Why the name:** the left side is a **product**, and the right side is a **sum** (or difference). Use these whenever a question gives a product of sines and cosines and you need to write it as a sum or difference.

**Memory aid from the lecture:**
- Mixed $\sin\cdot\cos$ gives $\sin$ terms on the right.
- $\cos\cdot\cos$ gives $\cos$ terms **added**.
- $\sin\cdot\sin$ gives $\cos$ terms, with $\cos(A-B)$ first and then **minus** $\cos(A+B)$.

```python
A, B = rng.uniform(-np.pi, np.pi, (2, 100_000))
verify("2 sinA cosB = sin(A+B) + sin(A−B)", 2*np.sin(A)*np.cos(B), np.sin(A + B) + np.sin(A - B))
verify("2 cosA sinB = sin(A+B) − sin(A−B)", 2*np.cos(A)*np.sin(B), np.sin(A + B) - np.sin(A - B))
verify("2 cosA cosB = cos(A+B) + cos(A−B)", 2*np.cos(A)*np.cos(B), np.cos(A + B) + np.cos(A - B))
verify("2 sinA sinB = cos(A−B) − cos(A+B)", 2*np.sin(A)*np.sin(B), np.cos(A - B) - np.cos(A + B))
```

### What product-to-sum looks like *(extra: a signal-processing view)*
Multiplying a fast wave $\sin(7t)$ by a slow wave $\cos(t)$ gives the same curve as **adding two waves** with frequencies $7 + 1 = 8$ and $7 - 1 = 6$:
$$2\sin 7t\cos t = \sin 8t + \sin 6t$$
This is the idea behind amplitude modulation (AM radio) and acoustic "beats". The slow wave forms an *envelope* around the fast one.

```python
t = np.linspace(0, 2*np.pi, 2000)
prod = 2*np.sin(7*t)*np.cos(t)
fig, axes = plt.subplots(2, 1, figsize=(11, 5.4), sharex=True)
ax = axes[0]
ax.plot(t, np.sin(8*t), color=AQUA, lw=1.2, alpha=0.9, label="sin 8t")
ax.plot(t, np.sin(6*t), color=VIOLET, lw=1.2, alpha=0.9, label="sin 6t")
ax.set_ylim(-1.4, 1.4); ax.legend(frameon=False, ncol=2, loc="upper right", fontsize=9)
ax.set_title("Two waves of nearby frequencies …", fontsize=11)
ax = axes[1]
ax.plot(t, np.sin(8*t) + np.sin(6*t), color=BLUE, lw=5, alpha=0.35, label="sin 8t + sin 6t  (sum)")
ax.plot(t, prod, color=INK, lw=1.2, label="2 sin 7t · cos t  (product)")
ax.plot(t, 2*np.cos(t), color=ORANGE, lw=1.4, ls="--", label="envelope ±2 cos t")
ax.plot(t, -2*np.cos(t), color=ORANGE, lw=1.4, ls="--")
ax.set_ylim(-2.4, 2.4); ax.legend(frameon=False, ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.22), fontsize=9)
ax.set_xticks(np.arange(0, 2.01, 0.5)*np.pi, ["0", "π/2", "π", "3π/2", "2π"]); ax.set_xlabel("t")
ax.set_title("… add up to exactly the product of a fast wave and a slow envelope", fontsize=11)
plt.tight_layout(); plt.show()
```

## 3. Showpiece: $\cos 20^\circ\cos 40^\circ\cos 60^\circ\cos 80^\circ = \dfrac{1}{16}$

(The lecturer will show a second, faster method in a later lecture. This one uses only what we know so far.)

**Facts used:**
- $2\cos A\cos B = \cos(A+B) + \cos(A-B)$
- $\cos 60^\circ = \frac12$
- $\cos 80^\circ = \cos(90^\circ - 10^\circ) = \sin 10^\circ$
- $\cos 100^\circ = \cos(90^\circ + 10^\circ) = -\sin 10^\circ$ (quadrant II, so $\cos < 0$)

**Step 1.** Replace $\cos 60^\circ$ by $\frac12$, then multiply and divide by 2 to create a $2\cos\cos$:
$$\text{LHS} = \frac12\cos 20^\circ\cos 40^\circ\cos 80^\circ = \frac14\,(2\cos 40^\circ\cos 20^\circ)\cos 80^\circ$$

**Step 2.** Product to sum: $2\cos 40^\circ\cos 20^\circ = \cos 60^\circ + \cos 20^\circ = \frac12 + \cos 20^\circ$
$$= \frac14\left(\frac12 + \cos 20^\circ\right)\cos 80^\circ = \frac18\cos 80^\circ + \frac14\cos 20^\circ\cos 80^\circ$$

**Step 3.** Again multiply and divide by 2: $\frac14\cos 80^\circ\cos 20^\circ = \frac18(2\cos 80^\circ\cos 20^\circ) = \frac18(\cos 100^\circ + \cos 60^\circ)$
$$= \frac18\cos 80^\circ + \frac18\cos 100^\circ + \frac18\cdot\frac12$$

**Step 4.** Allied angles: $\cos 80^\circ = \sin 10^\circ$ and $\cos 100^\circ = -\sin 10^\circ$, so those two terms cancel:
$$= \frac18\sin 10^\circ - \frac18\sin 10^\circ + \frac{1}{16} = \boxed{\frac{1}{16}} \quad\blacksquare$$

```python
c = lambda deg: np.cos(d(deg))
steps = [
    ("cos20·cos40·cos60·cos80",              c(20)*c(40)*c(60)*c(80)),
    ("¼·(2cos40·cos20)·cos80",               0.25*(2*c(40)*c(20))*c(80)),
    ("⅛·cos80 + ¼·cos20·cos80",              c(80)/8 + c(20)*c(80)/4),
    ("⅛·cos80 + ⅛·cos100 + 1/16",            c(80)/8 + c(100)/8 + 1/16),
    ("⅛·sin10 − ⅛·sin10 + 1/16",             np.sin(d(10))/8 - np.sin(d(10))/8 + 1/16),
]
for label, v in steps:
    print(f"{label:<32} = {v:.10f}")
print(f"{'1/16':<32} = {1/16:.10f}")
```

```python
# Extra: the product cos20·cos40·cos60·cos80 seen as bars, each factor shrinking the running product
factors = [20, 40, 60, 80]
running = np.cumprod([c(a) for a in factors])
fig, ax = plt.subplots(figsize=(8, 3.6))
bars = ax.bar(range(4), running, color=[BLUE, BLUE, ORANGE, BLUE], width=0.6)
for i, (a, r) in enumerate(zip(factors, running)):
    ax.text(i, r + 0.02, f"{r:.4f}", ha="center", va="bottom", fontsize=9)
ax.axhline(1/16, color=INK2, lw=1, ls="--"); ax.text(3.35, 1/16 + 0.01, "1/16", color=INK2, fontsize=9)
ax.set_xticks(range(4), ["× cos20°", "× cos40°", "× cos60°\n(= ½)", "× cos80°"])
ax.set_ylim(0, 1.05); ax.set_ylabel("running product")
ax.set_title("Running product, starting from 1: it ends at exactly 1/16 = 0.0625", fontsize=11)
plt.tight_layout(); plt.show()
```

## 4. Summary

**Techniques practised**
1. Expand with the formulas, then aim for the form of the RHS. Convert $\cos^2 \leftrightarrow 1 - \sin^2$ as needed.
2. Recognise an expanded pattern and **collapse** it: $\sin x\cos y + \cos x\sin y = \sin(x+y)$.
3. Split a fraction over its denominator so the terms telescope (Q4).
4. For conditions like $A+B+C=\pi$ or $x + 2x = 3x$, isolate one angle, take $\tan$ of both sides and expand.

**Results worth memorising (especially for JEE)**
| Result | Condition |
|---|---|
| $\sin(A+B)\sin(A-B) = \sin^2A - \sin^2B$ | always |
| $\cos(A+B)\cos(A-B) = \cos^2A - \sin^2B$ | always |
| $\tan A + \tan B + \tan C = \tan A\tan B\tan C$ | $A + B + C = \pi$ |
| $\tan 3x\tan 2x\tan x = \tan 3x - \tan 2x - \tan x$ | always (where defined) |
