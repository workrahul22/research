```python
import numpy as np
import matplotlib.pyplot as plt

# Same look as the earlier lectures
BLUE, ORANGE, AQUA, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#4a3aa7"
INK, INK2, GRID = "#0b0b0b", "#52514e", "#e4e3df"
plt.rcParams.update({
    "figure.dpi": 110, "axes.edgecolor": INK2, "axes.labelcolor": INK,
    "text.color": INK, "xtick.color": INK2, "ytick.color": INK2,
    "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.8,
    "axes.spines.top": False, "axes.spines.right": False, "font.size": 10,
})
rng = np.random.default_rng(10)

def sine_series(a, d, n):
    # Brute force: sin a + sin(a+d) + … + sin(a+(n−1)d)
    return sum(np.sin(a + k*d) for k in range(n))

def cosine_series(a, d, n):
    return sum(np.cos(a + k*d) for k in range(n))

def closed_form(f, a, d, n):
    # sin(n·d/2)/sin(d/2) × f((first + last)/2)
    return np.sin(n*d/2)/np.sin(d/2)*f(a + (n - 1)*d/2)
print("ready")
```

## 1. What Is a Trigonometric Series?

A **series** is a sum of terms that follow a pattern. In a **trigonometric series** the terms are trigonometric. Mathematics can only sum series with a **special pattern**; you can't write down an arbitrary sum and expect a closed form. (General series are covered in the *Sequences & Series* chapter.)

In this chapter there are three kinds:

| Type | Looks like |
|---|---|
| **Sine series** | $\sin\alpha + \sin(\alpha + \beta) + \sin(\alpha + 2\beta) + \cdots$ |
| **Cosine series** | $\cos\alpha + \cos(\alpha + \beta) + \cos(\alpha + 2\beta) + \cdots$ |
| **Miscellaneous** | no fixed look, but the same core idea (below) |

For the sine and cosine series, the **angles are in AP**: start at $\alpha$ and add the same **common difference** $\beta$ each time. With $n$ terms, the last angle is $\alpha + (n-1)\beta$.

### The master idea: split every term into a difference
> **Split each term into a difference of two terms, arranged so that consecutive terms cancel.**

If $T_r = f(r) - f(r+1)$, then
$$T_1 + T_2 + \cdots + T_n = \big(f(1) - f(2)\big) + \big(f(2) - f(3)\big) + \cdots + \big(f(n) - f(n+1)\big) = f(1) - f(n+1)$$
Everything in the middle cancels. This is called **telescoping**. The lecturer calls the cancellation **"loving"**: *if the terms don't cancel, the series won't sum.* Remember this for every series question.

---
## 2. Sum of the Sine Series

$$S = \sin\alpha + \sin(\alpha + \beta) + \sin(\alpha + 2\beta) + \cdots + \sin\big(\alpha + (n-1)\beta\big)$$

**Step 1: create a product-to-sum formula.** Multiply both sides by $2\sin\frac{\beta}{2}$. The **2** is needed to make the formula appear:
$$2\sin\tfrac{\beta}{2}\,S = 2\sin\alpha\sin\tfrac{\beta}{2} + 2\sin(\alpha + \beta)\sin\tfrac{\beta}{2} + \cdots + 2\sin\big(\alpha + (n-1)\beta\big)\sin\tfrac{\beta}{2}$$

**Step 2: apply $2\sin A\sin B = \cos(A - B) - \cos(A + B)$ to every term.** Each term becomes a difference:
$$\begin{aligned}
2\sin\alpha\sin\tfrac{\beta}{2} &= \cos\left(\alpha - \tfrac{\beta}{2}\right) - \cos\left(\alpha + \tfrac{\beta}{2}\right)\\
2\sin(\alpha+\beta)\sin\tfrac{\beta}{2} &= \cos\left(\alpha + \tfrac{\beta}{2}\right) - \cos\left(\alpha + \tfrac{3\beta}{2}\right)\\
2\sin(\alpha+2\beta)\sin\tfrac{\beta}{2} &= \cos\left(\alpha + \tfrac{3\beta}{2}\right) - \cos\left(\alpha + \tfrac{5\beta}{2}\right)\\
&\;\;\vdots\\
2\sin\big(\alpha + (n-1)\beta\big)\sin\tfrac{\beta}{2} &= \cos\left(\alpha + \tfrac{(2n-3)\beta}{2}\right) - \cos\left(\alpha + \tfrac{(2n-1)\beta}{2}\right)
\end{aligned}$$

**Step 3: telescope.** The negative part of each line cancels the positive part of the next. Only the first and last pieces survive:
$$2\sin\tfrac{\beta}{2}\,S = \cos\left(\alpha - \tfrac{\beta}{2}\right) - \cos\left(\alpha + \tfrac{(2n-1)\beta}{2}\right)$$

**Step 4: C & D**, $\cos C - \cos D = -2\sin\frac{C+D}{2}\sin\frac{C-D}{2}$:
- $\frac{C + D}{2} = \alpha + \frac{(n-1)\beta}{2}$
- $\frac{C - D}{2} = -\frac{n\beta}{2}$, and $\sin$ pushes the minus sign out, which cancels the leading minus

$$2\sin\tfrac{\beta}{2}\,S = 2\sin\left(\alpha + \tfrac{(n-1)\beta}{2}\right)\sin\tfrac{n\beta}{2}$$

$$\boxed{\;\sin\alpha + \sin(\alpha+\beta) + \cdots + \sin\big(\alpha + (n-1)\beta\big) = \frac{\sin\frac{n\beta}{2}}{\sin\frac{\beta}{2}}\;\sin\left(\alpha + \frac{(n-1)\beta}{2}\right)\;}$$

### Memorise it the lecturer's way
$$S_{\sin} = \frac{\sin\left(\dfrac{n\cdot\text{difference}}{2}\right)}{\sin\left(\dfrac{\text{difference}}{2}\right)}\times\sin\left(\frac{\text{first angle} + \text{last angle}}{2}\right)$$

Here $n$ is the number of terms, the difference is $\beta$, and the last angle is $\alpha + (n-1)\beta$, so $\frac{\text{first} + \text{last}}{2} = \alpha + \frac{(n-1)\beta}{2}$.

## 3. Sum of the Cosine Series

Exactly the same method gives the same formula, **with only the last $\sin$ changed to $\cos$**:

$$\boxed{\;\cos\alpha + \cos(\alpha+\beta) + \cdots + \cos\big(\alpha + (n-1)\beta\big) = \frac{\sin\frac{n\beta}{2}}{\sin\frac{\beta}{2}}\;\cos\left(\alpha + \frac{(n-1)\beta}{2}\right)\;}$$

$$S_{\cos} = \frac{\sin\left(\dfrac{n\cdot\text{difference}}{2}\right)}{\sin\left(\dfrac{\text{difference}}{2}\right)}\times\cos\left(\frac{\text{first angle} + \text{last angle}}{2}\right)$$

> **Homework from the lecture:** derive this yourself. Multiply by $2\sin\frac{\beta}{2}$, use $2\cos A\sin B = \sin(A+B) - \sin(A-B)$, telescope, then apply C & D.

```python
# Check both closed forms against brute-force sums for random α, β and n
ok_s = ok_c = 0; trials = 5000
for _ in range(trials):
    a, b, n = rng.uniform(-3, 3), rng.uniform(0.05, 3), rng.integers(1, 40)
    ok_s += np.isclose(sine_series(a, b, n), closed_form(np.sin, a, b, n))
    ok_c += np.isclose(cosine_series(a, b, n), closed_form(np.cos, a, b, n))
print(f"sine series formula:   {ok_s}/{trials} random (α, β, n) agree")
print(f"cosine series formula: {ok_c}/{trials} random (α, β, n) agree")
```

### Why the formula looks like this: phasors *(extra)*
Read $\cos\theta_k + i\sin\theta_k$ as a **unit arrow at angle $\theta_k$**. The cosine series and the sine series are the $x$- and $y$-components of the sum of $n$ unit arrows at angles $\alpha, \alpha+\beta, \dots$.

Placed head to tail, each arrow turns by $\beta$, so they trace part of a **regular polygon** inscribed in a circle. The resultant:
- has **length** $\left|\dfrac{\sin(n\beta/2)}{\sin(\beta/2)}\right|$, from the chord geometry of that circle
- points along the **average angle** $\alpha + \frac{(n-1)\beta}{2}$, which is $\frac{\text{first} + \text{last}}{2}$

Taking its $x$- and $y$-components gives exactly $S_{\cos}$ and $S_{\sin}$.

This is the same calculation behind **diffraction gratings** and the **Dirichlet kernel** in Fourier analysis. The factor $\frac{\sin(n\beta/2)}{\sin(\beta/2)}$ is the familiar "array factor" that peaks when all the arrows line up ($\beta \to 0$).

```python
a, b, n = np.deg2rad(10), np.deg2rad(28), 9
angles = a + b*np.arange(n)
pts = np.vstack([[0, 0], np.cumsum(np.c_[np.cos(angles), np.sin(angles)], axis=0)])
R = pts[-1]

fig, axes = plt.subplots(1, 2, figsize=(13, 5.2), gridspec_kw={"width_ratios": [1, 1.3]})
ax = axes[0]; ax.set_aspect("equal"); ax.grid(False)
for k in range(n):
    ax.annotate("", xy=tuple(pts[k + 1]), xytext=tuple(pts[k]),
                arrowprops=dict(arrowstyle="-|>", color=BLUE, lw=1.8))
ax.annotate("", xy=tuple(R), xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=ORANGE, lw=2.6))
ax.plot([R[0], R[0]], [0, R[1]], color=VIOLET, ls="--", lw=1); ax.plot([0, R[0]], [R[1], R[1]], color=VIOLET, ls="--", lw=1)
ax.text(R[0] + 0.1, R[1]/2, "S_sin", color=VIOLET, fontsize=10)
ax.text(R[0]/2, R[1] + 0.12, "S_cos", color=VIOLET, fontsize=10, ha="center")
ax.axhline(0, color=INK2, lw=0.8); ax.axvline(0, color=INK2, lw=0.8)
ax.set_title(f"{n} unit arrows, each turned by β = 28°:\nthe orange resultant = (S_cos, S_sin)", fontsize=10)
pad = 0.4
ax.set_xlim(pts[:, 0].min() - pad, pts[:, 0].max() + pad); ax.set_ylim(pts[:, 1].min() - pad, pts[:, 1].max() + pad)

ax = axes[1]
bb = np.linspace(-2*np.pi, 2*np.pi, 4001); bb = bb[np.abs(np.sin(bb/2)) > 1e-9]
for m, col in [(4, AQUA), (9, BLUE)]:
    ax.plot(bb, np.abs(np.sin(m*bb/2)/np.sin(bb/2)), color=col, lw=1.6, label=f"n = {m}")
ax.axvline(b, color=ORANGE, lw=1, ls=":"); ax.text(b + 0.08, 8.3, "β = 28°", color=ORANGE, fontsize=9)
ax.plot(b, abs(np.sin(n*b/2)/np.sin(b/2)), "o", color=ORANGE, ms=7, mec="white", mew=1.5, zorder=5)
ax.set_xticks(np.arange(-2, 2.01, 1)*np.pi, ["−2π", "−π", "0", "π", "2π"]); ax.set_xlabel("β (common difference)")
ax.set_ylabel("|sin(nβ/2) / sin(β/2)|"); ax.legend(frameon=True, framealpha=1, edgecolor="none", fontsize=9, loc="upper left", bbox_to_anchor=(0.08, 1))
ax.set_title("Length of the resultant: peaks at n when all arrows align (β = 0, ±2π)", fontsize=10)
plt.tight_layout(); plt.show()

print(f"|resultant| = {np.hypot(*R):.6f}   formula |sin(nβ/2)/sin(β/2)| = {abs(np.sin(n*b/2)/np.sin(b/2)):.6f}")
print(f"direction   = {np.rad2deg(np.arctan2(R[1], R[0])):.4f}°   (first + last)/2 = {np.rad2deg(a + (n-1)*b/2):.4f}°")
```

## 4. Q1: Find $\cos\frac{\pi}{11} + \cos\frac{3\pi}{11} + \cos\frac{5\pi}{11} + \cos\frac{7\pi}{11} + \cos\frac{9\pi}{11}$

**Identify the series:** a cosine series whose angles are in AP.
- Number of terms $n = 5$
- Common difference $\beta = \frac{2\pi}{11}$
- First angle $\frac{\pi}{11}$, last angle $\frac{9\pi}{11}$, so $\frac{\text{first} + \text{last}}{2} = \frac{5\pi}{11}$

Apply the formula:
$$S = \frac{\sin\left(\frac{5}{2}\cdot\frac{2\pi}{11}\right)}{\sin\frac{\pi}{11}}\cos\frac{5\pi}{11} = \frac{\sin\frac{5\pi}{11}\cos\frac{5\pi}{11}}{\sin\frac{\pi}{11}}$$

Multiply and divide by 2 to make $2\sin x\cos x = \sin 2x$:
$$= \frac{\sin\frac{10\pi}{11}}{2\sin\frac{\pi}{11}}$$

Allied angles: $\sin\frac{10\pi}{11} = \sin\left(\pi - \frac{\pi}{11}\right) = \sin\frac{\pi}{11}$, so they cancel:
$$S = \boxed{\frac12}$$

```python
k = np.array([1, 3, 5, 7, 9])
vals = np.cos(k*np.pi/11)
print("terms:", np.round(vals, 6), "  sum =", round(vals.sum(), 12))
print("formula:", closed_form(np.cos, np.pi/11, 2*np.pi/11, 5))

fig, ax = plt.subplots(figsize=(8.5, 3.4))
cum = np.cumsum(vals)
ax.bar(range(5), vals, color=[BLUE if v > 0 else ORANGE for v in vals], width=0.55, label="term")
ax.plot(range(5), cum, "o-", color=INK, lw=1.5, label="running sum")
ax.axhline(0.5, color=VIOLET, lw=1, ls="--"); ax.text(4.3, 0.53, "½", color=VIOLET, fontsize=11)
ax.axhline(0, color=INK2, lw=1)
ax.set_xticks(range(5), [f"cos({j}π/11)" for j in k])
ax.legend(frameon=False, fontsize=9, loc="lower left")
ax.set_title("Q1: five cosines in AP sum to exactly ½", fontsize=11)
plt.tight_layout(); plt.show()
```

## 5. Q2: Show $\displaystyle\sum_{r=1}^{n-1}\sin^2\frac{r\pi}{n} = \frac{n}{2}$

**Reading $\Sigma$:** "sum, for $r = 1, 2, \dots, n-1$, of $\sin^2\frac{r\pi}{n}$". Put $r = 1$, then $r = 2$, and so on up to $r = n-1$, and add the results. (The lecturer promises a dedicated lecture on $\Sigma$ notation.)

**Step 1: remove the square.** The series formulas need plain $\sin$/$\cos$, not powers. Use $\sin^2 x = \frac{1 - \cos 2x}{2}$:
$$\sum_{r=1}^{n-1}\sin^2\frac{r\pi}{n} = \frac12\sum_{r=1}^{n-1}1 \;-\; \frac12\sum_{r=1}^{n-1}\cos\frac{2r\pi}{n} = \frac{n-1}{2} - \frac12\,C$$
($\sum 1$ over $n - 1$ values of $r$ is just $n - 1$.)

**Step 2: $C = \cos\frac{2\pi}{n} + \cos\frac{4\pi}{n} + \cdots + \cos\frac{2(n-1)\pi}{n}$ is a cosine series.** It has $n - 1$ terms, difference $\frac{2\pi}{n}$, first angle $\frac{2\pi}{n}$ and last angle $\frac{2(n-1)\pi}{n}$, so $\frac{\text{first} + \text{last}}{2} = \pi$:
$$C = \frac{\sin\frac{(n-1)\pi}{n}}{\sin\frac{\pi}{n}}\cdot\cos\pi = \frac{\sin\left(\pi - \frac{\pi}{n}\right)}{\sin\frac{\pi}{n}}\cdot(-1) = -1$$

**Step 3: combine.**
$$\sum_{r=1}^{n-1}\sin^2\frac{r\pi}{n} = \frac{n-1}{2} + \frac12 = \boxed{\frac{n}{2}}$$

```python
ns = np.arange(2, 31)
sums = [np.sum(np.sin(np.arange(1, n)*np.pi/n)**2) for n in ns]
print("n:      ", ns[:10])
print("Σ sin²: ", np.round(sums[:10], 10))
print("n/2:    ", ns[:10]/2)

fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
ax = axes[0]
ax.plot(ns, sums, "o", color=BLUE, ms=6, mec="white", mew=1.2, label="Σ sin²(rπ/n), computed")
ax.plot(ns, ns/2, color=ORANGE, lw=1.3, ls="--", label="n/2")
ax.set_xlabel("n"); ax.legend(frameon=False, fontsize=9)
ax.set_title("Q2: the sum equals n/2 for every n", fontsize=11)

ax = axes[1]
n = 12; r = np.arange(1, n)
ax.bar(r, np.sin(r*np.pi/n)**2, color=BLUE, width=0.6, label="sin²(rπ/n)")
ax.bar(r, np.cos(r*np.pi/n)**2, bottom=np.sin(r*np.pi/n)**2, color=ORANGE, alpha=0.5, width=0.6, label="cos²(rπ/n)")
ax.set_xticks(r); ax.set_xlabel(f"r   (n = {n})"); ax.set_ylim(0, 1.15)
ax.legend(frameon=False, fontsize=9, ncol=2, loc="upper center")
ax.set_title("sin² + cos² = 1 in every bar, and the sin² bars are symmetric about r = n/2", fontsize=10)
plt.tight_layout(); plt.show()
```

## 6. Summary

**Concept:** sum a trigonometric series by **splitting each term into a difference** so the terms cancel (telescoping). In the lecturer's words, *the cancellation is the "loving" part*.

**The two master formulas** (angles $\alpha, \alpha + \beta, \dots, \alpha + (n-1)\beta$):
$$\sum_{k=0}^{n-1}\sin(\alpha + k\beta) = \frac{\sin\frac{n\beta}{2}}{\sin\frac{\beta}{2}}\,\sin\left(\alpha + \frac{(n-1)\beta}{2}\right), \qquad
\sum_{k=0}^{n-1}\cos(\alpha + k\beta) = \frac{\sin\frac{n\beta}{2}}{\sin\frac{\beta}{2}}\,\cos\left(\alpha + \frac{(n-1)\beta}{2}\right)$$

In words: $\dfrac{\sin(n \cdot \text{diff}/2)}{\sin(\text{diff}/2)} \times \sin$ or $\cos\left(\dfrac{\text{first} + \text{last}}{2}\right)$.

**How to derive them:** multiply by $2\sin\frac{\beta}{2}$, apply product-to-sum to every term, telescope, then use C & D.

**Problem technique:**
1. Check that the angles are in AP. Identify $n$, the difference, the first angle and the last angle.
2. Remove powers first ($\sin^2 x = \frac{1 - \cos 2x}{2}$, and so on).
3. Finish with $2\sin x\cos x = \sin 2x$ and allied-angle simplifications.
