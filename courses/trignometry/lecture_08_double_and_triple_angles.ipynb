## 1. $\sin 2\theta$

### Form 1 (from Lecture 7)
$$\boxed{\sin 2\theta = 2\sin\theta\cos\theta} \qquad \text{masala form: } \sin(\text{m}) = 2\sin\tfrac{\text{m}}{2}\cos\tfrac{\text{m}}{2}$$

### Form 2: in terms of $\tan\theta$
Any number can be written over 1, and $1 = \sin^2\theta + \cos^2\theta$:
$$\sin 2\theta = \frac{2\sin\theta\cos\theta}{1} = \frac{2\sin\theta\cos\theta}{\sin^2\theta + \cos^2\theta}$$
Divide the numerator and denominator by $\cos^2\theta$:
$$\boxed{\sin 2\theta = \frac{2\tan\theta}{1 + \tan^2\theta}} \qquad \text{masala form: } \sin(\text{m}) = \frac{2\tan\frac{\text{m}}{2}}{1 + \tan^2\frac{\text{m}}{2}}$$

**Example:** $\sin\theta = \dfrac{2\tan\frac{\theta}{2}}{1 + \tan^2\frac{\theta}{2}}$. Halve whatever is inside and put it on the right.

## 2. $\cos 2\theta$

Put $A = B = \theta$ in $\cos(A+B) = \cos A\cos B - \sin A\sin B$:
$$\boxed{\cos 2\theta = \cos^2\theta - \sin^2\theta} \qquad (1)$$
This is the **root** of every other $\cos 2\theta$ formula.

Replace $\cos^2\theta = 1 - \sin^2\theta$:
$$\boxed{\cos 2\theta = 1 - 2\sin^2\theta} \qquad (2)$$

Replace $\sin^2\theta = 1 - \cos^2\theta$ instead:
$$\boxed{\cos 2\theta = 2\cos^2\theta - 1} \qquad (3)$$

Write $(1)$ over $1 = \cos^2\theta + \sin^2\theta$ and divide by $\cos^2\theta$:
$$\boxed{\cos 2\theta = \frac{1 - \tan^2\theta}{1 + \tan^2\theta}} \qquad (4)$$

**Which one to use:**
| Formula | Use it when you want $\cos 2\theta$ in terms of … |
|---|---|
| $(1)$ $\cos^2\theta - \sin^2\theta$ | both $\sin$ and $\cos$ |
| $(2)$ $1 - 2\sin^2\theta$ | $\sin$ only |
| $(3)$ $2\cos^2\theta - 1$ | $\cos$ only |
| $(4)$ $\frac{1 - \tan^2\theta}{1 + \tan^2\theta}$ | $\tan$ only |

**Masala form:** $\cos(\text{m}) = \cos^2\frac{\text{m}}{2} - \sin^2\frac{\text{m}}{2} = 1 - 2\sin^2\frac{\text{m}}{2} = 2\cos^2\frac{\text{m}}{2} - 1 = \dfrac{1 - \tan^2\frac{\text{m}}{2}}{1 + \tan^2\frac{\text{m}}{2}}$

### 2.1 Power reduction: $\sin^2\theta$, $\cos^2\theta$, $\tan^2\theta$
Rearrange $(2)$ and $(3)$:
$$\boxed{\sin^2\theta = \frac{1 - \cos 2\theta}{2}}, \qquad \boxed{\cos^2\theta = \frac{1 + \cos 2\theta}{2}}$$
Dividing one by the other:
$$\boxed{\tan^2\theta = \frac{1 - \cos 2\theta}{1 + \cos 2\theta}}$$

These are used constantly, in trigonometry and also in **calculus**, for example to integrate $\sin^2 x$. Equivalent forms are $1 - \cos 2\theta = 2\sin^2\theta$ and $1 + \cos 2\theta = 2\cos^2\theta$.

**Why "power reduction":** a **square** of a trig function becomes a **first power** of $\cos$ at double the angle.

```python
t = np.linspace(0, 2*np.pi, 1000)
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
ax = axes[0]
ax.plot(t, np.cos(2*t), color=INK2, lw=1.2, ls=":", label="cos 2θ")
ax.plot(t, np.sin(t)**2, color=BLUE, lw=5, alpha=0.35, label="sin²θ")
ax.plot(t, (1 - np.cos(2*t))/2, color=BLUE, lw=1.4, ls="--", label="(1 − cos 2θ)/2")
ax.plot(t, np.cos(t)**2, color=ORANGE, lw=5, alpha=0.35, label="cos²θ")
ax.plot(t, (1 + np.cos(2*t))/2, color=ORANGE, lw=1.4, ls="--", label="(1 + cos 2θ)/2")
ax.axhline(0.5, color=INK2, lw=0.8)
ax.set_xticks(*PI_TICKS); ax.set_ylim(-1.15, 1.15); ax.set_xlabel("θ")
ax.legend(frameon=False, ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.2), fontsize=8.5)
ax.set_title("sin²θ and cos²θ are cos 2θ shrunk by ½ and shifted to centre on ½", fontsize=10.5)

ax = axes[1]
ax.plot(t, np.sin(t)**2 + np.cos(t)**2, color=INK, lw=2, label="sin²θ + cos²θ = 1")
ax.fill_between(t, 0, np.sin(t)**2, color=BLUE, alpha=0.5, lw=0, label="sin²θ")
ax.fill_between(t, np.sin(t)**2, 1, color=ORANGE, alpha=0.5, lw=0, label="cos²θ")
ax.set_xticks(*PI_TICKS); ax.set_ylim(0, 1.1); ax.set_xlabel("θ")
ax.legend(frameon=False, ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.2), fontsize=8.5)
ax.set_title("Both oscillate at double frequency and always sum to 1", fontsize=10.5)
plt.tight_layout(); plt.show()
```

## 3. $\tan 2\theta$

Put $A = B = \theta$ in $\tan(A+B) = \dfrac{\tan A + \tan B}{1 - \tan A\tan B}$:
$$\boxed{\tan 2\theta = \frac{2\tan\theta}{1 - \tan^2\theta}} \qquad \text{masala form: } \tan(\text{m}) = \frac{2\tan\frac{\text{m}}{2}}{1 - \tan^2\frac{\text{m}}{2}}$$
There is just this one formula.

**Notice the pattern in the $\tan$-forms:**
$$\sin 2\theta = \frac{2\tan\theta}{1 + \tan^2\theta}, \qquad \cos 2\theta = \frac{1 - \tan^2\theta}{1 + \tan^2\theta}, \qquad \tan 2\theta = \frac{2\tan\theta}{1 - \tan^2\theta}$$
The denominators are $1 + \tan^2$, $1 + \tan^2$ and $1 - \tan^2$. Also, $\tan 2\theta = \dfrac{\sin 2\theta}{\cos 2\theta}$ as expected.

**What about $\cot 2\theta$, $\sec 2\theta$, $\csc 2\theta$?** No new formulas are needed. Use the reciprocals:
$$\cot 2\theta = \frac{1}{\tan 2\theta}, \qquad \sec 2\theta = \frac{1}{\cos 2\theta}, \qquad \csc 2\theta = \frac{1}{\sin 2\theta}$$

### A picture of the $\tan$-forms *(extra)*
Let $t = \tan\theta$. All three $\tan$-forms come from **one right triangle** with angle $2\theta$ whose sides are
$$\text{base} = 1 - t^2, \qquad \text{perpendicular} = 2t, \qquad \text{hypotenuse} = 1 + t^2$$
It really is a right triangle: $(1 - t^2)^2 + (2t)^2 = 1 + 2t^2 + t^4 = (1 + t^2)^2$. Reading off $P/H$, $B/H$ and $P/B$ gives $\sin 2\theta$, $\cos 2\theta$ and $\tan 2\theta$.

(Choosing rational $t$ generates Pythagorean triples: $t = \frac12$ gives $3$-$4$-$5$ after scaling. In calculus this substitution is the "$t = \tan\frac{x}{2}$" trick.)

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
for ax, tt in zip(axes, [1/3, 1/2, 2/3]):
    B_, P_, H_ = 1 - tt**2, 2*tt, 1 + tt**2
    ax.set_aspect("equal"); ax.axis("off")
    ax.add_patch(Polygon([(0, 0), (B_, 0), (B_, P_)], closed=True, fc=BLUE + "1f", ec=INK, lw=2))
    ang = np.rad2deg(np.arctan2(P_, B_))
    ax.add_patch(Arc((0, 0), 0.45, 0.45, theta1=0, theta2=ang, color=VIOLET, lw=2))
    ax.text(0.27, 0.05, "2θ", color=VIOLET, fontsize=11)
    ax.text(B_/2, -0.05, f"1 − t² = {B_:.3f}", ha="center", va="top", fontsize=9)
    ax.text(B_ + 0.03, P_/2, f"2t = {P_:.3f}", va="center", fontsize=9)
    ax.text(B_/2 - 0.05, P_/2 + 0.05, f"1 + t² = {H_:.3f}", ha="right", fontsize=9, rotation=ang)
    ax.set_xlim(-0.15, 1.45); ax.set_ylim(-0.25, 1.45)
    s2, c2, t2 = P_/H_, B_/H_, P_/B_
    ok = np.allclose([s2, c2, t2], [np.sin(2*np.arctan(tt)), np.cos(2*np.arctan(tt)), np.tan(2*np.arctan(tt))])
    ax.set_title(f"t = tan θ = {tt:.3f}\nsin2θ = {s2:.3f}, cos2θ = {c2:.3f}, tan2θ = {t2:.3f}  {'✓' if ok else '✗'}", fontsize=10)
fig.suptitle("One triangle for all three tan-forms: sides 1 − t², 2t, 1 + t²", fontsize=12, y=1.02)
plt.tight_layout(); plt.show()
```

```python
x = rng.uniform(-3, 3, 100_000); tn = np.tan(x)
print("All double-angle formulae, checked at 100,000 random θ:")
verify("sin2θ = 2 sinθ cosθ",               np.sin(2*x), 2*np.sin(x)*np.cos(x))
verify("sin2θ = 2tanθ/(1 + tan²θ)",          np.sin(2*x), 2*tn/(1 + tn**2))
verify("cos2θ = cos²θ − sin²θ",              np.cos(2*x), np.cos(x)**2 - np.sin(x)**2)
verify("cos2θ = 1 − 2sin²θ",                 np.cos(2*x), 1 - 2*np.sin(x)**2)
verify("cos2θ = 2cos²θ − 1",                 np.cos(2*x), 2*np.cos(x)**2 - 1)
verify("cos2θ = (1 − tan²θ)/(1 + tan²θ)",    np.cos(2*x), (1 - tn**2)/(1 + tn**2))
verify("tan2θ = 2tanθ/(1 − tan²θ)",          np.tan(2*x), 2*tn/(1 - tn**2))
verify("tan²θ = (1 − cos2θ)/(1 + cos2θ)",    tn**2, (1 - np.cos(2*x))/(1 + np.cos(2*x)))
```

## 4. Triple Angles

$$\boxed{\sin 3\theta = 3\sin\theta - 4\sin^3\theta}$$
$$\boxed{\cos 3\theta = 4\cos^3\theta - 3\cos\theta}$$
$$\boxed{\tan 3\theta = \frac{3\tan\theta - \tan^3\theta}{1 - 3\tan^2\theta}}$$

**Memory aids (from the lecture):**
- $\sin 3\theta$: **3 then 4**, with the cube last. $\cos 3\theta$: **4 then 3**, with the cube first. Both have a minus sign in the middle.
- $\tan 3\theta$: three 3s. $3\tan$ on top, $1 - 3\tan^2$ below, and $\tan^3$ subtracted on top. Powers $1, 3$ on top and $0, 2$ below.

### Proof of $\sin 3\theta$ (done in class)
$$\sin 3\theta = \sin(2\theta + \theta) = \sin 2\theta\cos\theta + \cos 2\theta\sin\theta$$
Use $\sin 2\theta = 2\sin\theta\cos\theta$. For $\cos 2\theta$ pick the **sin-only** form $1 - 2\sin^2\theta$, because the target is in $\sin$:
$$= 2\sin\theta\cos^2\theta + (1 - 2\sin^2\theta)\sin\theta$$
Replace $\cos^2\theta = 1 - \sin^2\theta$:
$$= 2\sin\theta - 2\sin^3\theta + \sin\theta - 2\sin^3\theta = 3\sin\theta - 4\sin^3\theta \quad\blacksquare$$
("Two mangoes plus one mango is three mangoes; minus two bananas minus two bananas is minus four bananas.")

### Homework from the lecture: prove $\cos 3\theta$ and $\tan 3\theta$
<details>
<summary>Proofs (open after attempting)</summary>

**$\cos 3\theta$:** $\cos(2\theta + \theta) = \cos 2\theta\cos\theta - \sin 2\theta\sin\theta = (2\cos^2\theta - 1)\cos\theta - 2\sin^2\theta\cos\theta$
$= 2\cos^3\theta - \cos\theta - 2(1 - \cos^2\theta)\cos\theta = 4\cos^3\theta - 3\cos\theta$ $\blacksquare$

**$\tan 3\theta$:** with $t = \tan\theta$, $\tan(2\theta + \theta) = \dfrac{\tan 2\theta + t}{1 - t\tan 2\theta}$ with $\tan 2\theta = \dfrac{2t}{1 - t^2}$.
Numerator: $\dfrac{2t + t(1 - t^2)}{1 - t^2} = \dfrac{3t - t^3}{1 - t^2}$. Denominator: $\dfrac{(1 - t^2) - 2t^2}{1 - t^2} = \dfrac{1 - 3t^2}{1 - t^2}$.
Divide: $\tan 3\theta = \dfrac{3t - t^3}{1 - 3t^2}$ $\blacksquare$
</details>

```python
x = rng.uniform(-3, 3, 100_000); tn = np.tan(x)
verify("sin3θ = 3sinθ − 4sin³θ", np.sin(3*x), 3*np.sin(x) - 4*np.sin(x)**3)
verify("cos3θ = 4cos³θ − 3cosθ", np.cos(3*x), 4*np.cos(x)**3 - 3*np.cos(x))
verify("tan3θ = (3tanθ − tan³θ)/(1 − 3tan²θ)", np.tan(3*x), (3*tn - tn**3)/(1 - 3*tn**2))

t = np.linspace(0, 2*np.pi, 1000)
fig, ax = plt.subplots(figsize=(10, 3.6))
ax.plot(t, np.sin(3*t), color=BLUE, lw=5, alpha=0.35, label="sin 3θ")
ax.plot(t, 3*np.sin(t) - 4*np.sin(t)**3, color=BLUE, lw=1.3, ls="--", label="3 sinθ − 4 sin³θ")
ax.plot(t, 3*np.sin(t), color=ORANGE, lw=1.6, label="3 sinθ  (common mistake)")
ax.axhline(0, color=INK2, lw=1); ax.set_xticks(*PI_TICKS); ax.set_ylim(-3.3, 3.3); ax.set_xlabel("θ")
ax.legend(frameon=False, ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.2), fontsize=9)
ax.set_title("sin 3θ = 3 sinθ − 4 sin³θ  (the −4 sin³θ term keeps it between −1 and 1)", fontsize=11)
plt.tight_layout(); plt.show()
```

### Extra: $\cos n\theta$ as a polynomial in $\cos\theta$ (Chebyshev polynomials)
The formulas $\cos 2\theta = 2\cos^2\theta - 1$ and $\cos 3\theta = 4\cos^3\theta - 3\cos\theta$ say that $\cos n\theta$ is a **polynomial** in $c = \cos\theta$:
$$T_1(c) = c, \quad T_2(c) = 2c^2 - 1, \quad T_3(c) = 4c^3 - 3c, \quad T_{n+1}(c) = 2c\,T_n(c) - T_{n-1}(c)$$
These are the **Chebyshev polynomials**. They are used in numerical approximation, spectral methods and polynomial filters (including graph neural networks such as ChebNet). On $[-1, 1]$ they wiggle between $-1$ and $1$, exactly like $\cos n\theta$ does.

```python
c = np.linspace(-1, 1, 600)
T = [np.ones_like(c), c]
for n in range(2, 6):
    T.append(2*c*T[-1] - T[-2])                       # recurrence T_{n+1} = 2c·T_n − T_{n−1}
fig, ax = plt.subplots(figsize=(8, 4))
for n, col in zip([1, 2, 3, 4], [INK2, BLUE, ORANGE, AQUA]):
    ax.plot(c, T[n], color=col, lw=2, label=f"T{n}(c) = cos {n}θ")
ax.set_xlabel("c = cos θ"); ax.set_ylim(-1.15, 1.15)
ax.legend(frameon=False, ncol=4, loc="upper center", bbox_to_anchor=(0.5, -0.18), fontsize=9)
ax.set_title("cos nθ as a polynomial in cos θ:  T₂ = 2c² − 1,  T₃ = 4c³ − 3c", fontsize=11)
plt.tight_layout(); plt.show()

th = rng.uniform(0, np.pi, 50_000)
for n in range(2, 6):
    Tn = [np.ones_like(th), np.cos(th)]
    for _ in range(2, n + 1): Tn.append(2*np.cos(th)*Tn[-1] - Tn[-2])
    verify(f"T{n}(cos θ) = cos {n}θ", Tn[n], np.cos(n*th))
```

## 5. Worked Example: $\tan 22\frac12^\circ$

Use $\tan 2x = \dfrac{2\tan x}{1 - \tan^2 x}$ with $x = 22\frac12^\circ$, so that $2x = 45^\circ$. Let $t = \tan 22\frac12^\circ$:
$$\tan 45^\circ = 1 = \frac{2t}{1 - t^2} \;\Longrightarrow\; 1 - t^2 = 2t \;\Longrightarrow\; t^2 + 2t - 1 = 0$$
Quadratic formula:
$$t = \frac{-2 \pm \sqrt{4 + 4}}{2} = -1 \pm \sqrt2$$

**Reject $-1 - \sqrt2$.** It is negative, but $22\frac12^\circ$ is in quadrant I, where $\tan > 0$. So
$$\boxed{\tan 22\tfrac12^\circ = \sqrt2 - 1 \approx 0.414}, \qquad \cot 22\tfrac12^\circ = \frac{1}{\sqrt2 - 1} = \sqrt2 + 1$$

In radians, $22\frac12^\circ = \frac{\pi}{8}$. Since $\tan\left(\frac{\pi}{2} - x\right) = \cot x$:
$$\tan\frac{\pi}{8} = \cot\frac{3\pi}{8} = \sqrt2 - 1, \qquad \cot\frac{\pi}{8} = \tan\frac{3\pi}{8} = \sqrt2 + 1 \quad (67\tfrac12^\circ)$$

> **For objective exams (JEE, etc.):** memorise these two values.

```python
t1, t2 = -1 + np.sqrt(2), -1 - np.sqrt(2)
print(f"roots of t² + 2t − 1 = 0:  {t1:.6f} (keep: positive),  {t2:.6f} (reject: tan > 0 in QI)")
print(f"tan 22.5°  = {np.tan(d(22.5)):.6f},   √2 − 1 = {np.sqrt(2) - 1:.6f}")
print(f"cot 22.5°  = {1/np.tan(d(22.5)):.6f},   √2 + 1 = {np.sqrt(2) + 1:.6f}")
print(f"tan 67.5°  = {np.tan(d(67.5)):.6f}")

# Extra: a purely geometric picture. In a right isosceles triangle (1, 1, √2), extend the base by √2.
# The line from the far end to the apex makes 22.5° with the base, so tan 22.5° = 1/(1 + √2) = √2 − 1.
fig, ax = plt.subplots(figsize=(8, 3.6)); ax.set_aspect("equal"); ax.axis("off")
Bp, Cp, Ap, Dp = np.array([0, 0]), np.array([1, 0]), np.array([1, 1]), np.array([-np.sqrt(2), 0])
ax.add_patch(Polygon([Bp, Cp, Ap], closed=True, fc=BLUE + "1f", ec=INK, lw=2))
ax.add_patch(Polygon([Dp, Bp, Ap], closed=True, fc=ORANGE + "1f", ec=INK, lw=1.5))
ax.plot([0.9, 0.9, 1], [0, 0.1, 0.1], color=INK2, lw=1)
ax.add_patch(Arc(Bp, 0.5, 0.5, theta1=0, theta2=45, color=BLUE, lw=2)); ax.text(0.27, 0.07, "45°", color=BLUE, fontsize=10)
ax.add_patch(Arc(Dp, 1.2, 1.2, theta1=0, theta2=22.5, color=ORANGE, lw=2)); ax.text(Dp[0] + 0.65, 0.07, "22.5°", color=ORANGE, fontsize=10)
ax.text(-np.sqrt(2)/2, -0.06, "√2", ha="center", va="top", fontsize=11)
ax.text(0.5, -0.06, "1", ha="center", va="top", fontsize=11)
ax.text(1.04, 0.5, "1", va="center", fontsize=11)
ax.text(0.36, 0.56, "√2", fontsize=11, color=INK2, ha="right")
ax.set_xlim(-1.6, 1.3); ax.set_ylim(-0.25, 1.15)
ax.set_title("Extend the base by √2: an isosceles triangle with base angles 22.5°, so tan 22.5° = 1/(√2 + 1) = √2 − 1", fontsize=10)
plt.show()
```

## 7. Summary: All Multiple-Angle Formulae So Far

| | Formulae |
|---|---|
| $\sin 2\theta$ | $2\sin\theta\cos\theta = \dfrac{2\tan\theta}{1 + \tan^2\theta}$ |
| $\cos 2\theta$ | $\cos^2\theta - \sin^2\theta = 1 - 2\sin^2\theta = 2\cos^2\theta - 1 = \dfrac{1 - \tan^2\theta}{1 + \tan^2\theta}$ |
| $\tan 2\theta$ | $\dfrac{2\tan\theta}{1 - \tan^2\theta}$ |
| Power reduction | $\sin^2\theta = \frac{1 - \cos 2\theta}{2}$, $\cos^2\theta = \frac{1 + \cos 2\theta}{2}$, $\tan^2\theta = \frac{1 - \cos 2\theta}{1 + \cos 2\theta}$ |
| $\sin 3\theta$ | $3\sin\theta - 4\sin^3\theta$ |
| $\cos 3\theta$ | $4\cos^3\theta - 3\cos\theta$ |
| $\tan 3\theta$ | $\dfrac{3\tan\theta - \tan^3\theta}{1 - 3\tan^2\theta}$ |
| Values | $\tan\frac{\pi}{8} = \sqrt2 - 1$, $\cot\frac{\pi}{8} = \sqrt2 + 1$ |

**"Masala" rule:** each formula holds with any angle inside, as long as the angle on the right is the left angle divided appropriately (half for the $2\theta$ forms, a third for the $3\theta$ forms).
